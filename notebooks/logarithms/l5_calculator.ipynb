# Find log 7 the calculator's way

The natural log is the area under y = 1/x, and Mercator turned that area into a sum of fractions. Shrink 7 close to
1, add a few fractions, and change the base: that is log 7.

Press **Run** on each grey box, top to bottom (or Shift + Enter). Then change the numbers and run it again.

```python
# Setup: in the browser version (JupyterLite) the sliders need one install; elsewhere this does nothing.
import sys
if "pyodide" in sys.modules:
    import piplite
    await piplite.install(["ipywidgets"])
```

```python
import math
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider
```

## 1 · Equal ratios, equal areas

```python
def area(a=1.0, b=2.0, steps=100000):
    h = (b - a) / steps
    return sum(h / (a + (i + 0.5) * h) for i in range(steps))      # add up thin strips under 1/x

def show(start=1.0, ratio=2.0):
    end = start * ratio
    xs = [0.3 + 9.7 * i / 400 for i in range(401)]
    fig, ax = plt.subplots(figsize=(9, 3.2))
    ax.plot(xs, [1 / x for x in xs], color="tab:orange")
    fx = [start + (end - start) * i / 100 for i in range(101)]
    ax.fill_between(fx, [1 / x for x in fx], alpha=0.3, color="tab:orange")
    ax.set_ylim(0, 2.2); ax.set_title(f"area from {start:g} to {end:g} = {area(start, end, 20000):.5f}   (ln {ratio:g} = {math.log(ratio):.5f})")
    plt.show()

interact(show, start=FloatSlider(value=1, min=0.5, max=4, step=0.5), ratio=FloatSlider(value=2, min=1.1, max=4, step=0.1));
```

## 2 · Mercator's series: ln(1 + x) = x − x²/2 + x³/3 − …

```python
def mercator(x=0.1, terms=5):
    s = 0.0
    for n in range(1, terms + 1):
        s += (-1) ** (n + 1) * x ** n / n
        print(f"after {n} terms: {s:.10f}")
    print(f"true ln(1 + {x:g}) = {math.log(1 + x):.10f}")

interact(mercator, x=FloatSlider(value=0.1, min=-0.9, max=1.0, step=0.05), terms=IntSlider(value=5, min=1, max=40));
```

Small x: a handful of terms is enough. x near 1: it creeps. That's why a calculator first moves the number close
to 1.

## 3 · log 7, step by step

```python
ln2 = sum(0.5 ** n / n for n in range(1, 40))          # ln 2 = −ln(1 − ½)
ln0875 = -sum((1 / 8) ** n / n for n in range(1, 12))   # ln(1 − 1/8)
ln125 = sum(0.2 ** n / n for n in range(1, 20))         # ln 1.25 = −ln(1 − 0.2)
ln7 = 3 * ln2 + ln0875                                  # 7 = 2³ × 0.875
ln10 = 3 * ln2 + ln125                                  # 10 = 2³ × 1.25
print(f"ln 2 = {ln2:.6f}   ln 7 = {ln7:.6f}   ln 10 = {ln10:.6f}")
print(f"log 7 = ln 7 ÷ ln 10 = {ln7 / ln10:.6f}    (math.log10(7) = {math.log10(7):.6f})")
```

## 4 · Change of base

```python
def change_of_base(x=7.0, base=2.0):
    print(f"log base {base:g} of {x:g} = ln {x:g} ÷ ln {base:g} = {math.log(x) / math.log(base):.6f}")

interact(change_of_base, x=FloatSlider(value=7, min=0.1, max=1000, step=0.1), base=FloatSlider(value=2, min=1.1, max=20, step=0.1));
```
