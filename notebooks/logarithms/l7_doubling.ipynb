# How long until it doubles, or halves?

When the question is "how many times?", the answer is a logarithm: if bᵗ = N, then t = log N ÷ log b. Use it for
growth, for half-lives, and to date a bone.

Press **Run** on each grey box, top to bottom (or Shift + Enter). Then change the numbers and run it again.

```python
# Setup: in the browser version (JupyterLite) the sliders need one install; elsewhere this does nothing.
import sys
if "pyodide" in sys.modules:
    import piplite
    await piplite.install(["ipywidgets"])
```

```python
import math
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
```

## 1 · Solve bᵗ = N

```python
def solve(b=2.0, N=10.0):
    t = math.log10(N) / math.log10(b)
    print(f"t = log {N:g} ÷ log {b:g} = {t:.4f}      check: {b:g}^{t:.4f} = {b ** t:.4f}")

interact(solve, b=FloatSlider(value=2, min=1.01, max=10, step=0.01), N=FloatSlider(value=10, min=0.01, max=1000, step=0.01));
```

## 2 · An exponential straightens on a log axis

```python
def straighten(rate=7.0):
    ts = list(range(0, 51))
    ys = [(1 + rate / 100) ** t for t in ts]
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(10, 3.4))
    a1.plot(ts, ys); a1.set_title(f"+{rate:g} % a step, ordinary axis")
    a2.semilogy(ts, ys); a2.set_title("log axis: a straight line")
    for a in (a1, a2): a.grid(alpha=0.3); a.set_xlabel("steps")
    plt.show()
    print(f"doubling time: log 2 ÷ log {1 + rate / 100:g} = {math.log10(2) / math.log10(1 + rate / 100):.2f} steps")

interact(straighten, rate=FloatSlider(value=7, min=0.5, max=50, step=0.5));
```

## 3 · Date a bone with carbon-14 (half-life about 5,700 years)

```python
def carbon(percent_left=10.0, half_life=5700.0):
    n = math.log10(percent_left / 100) / math.log10(0.5)
    print(f"{percent_left:g} % left = (½)^n  →  n = {n:.2f} half-lives  →  about {n * half_life:,.0f} years")

interact(carbon, percent_left=FloatSlider(value=10, min=0.5, max=99, step=0.5), half_life=FloatSlider(value=5700, min=100, max=10000, step=100));
```
