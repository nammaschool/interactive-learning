# Multiply by adding: the table of twos

In the video we multiplied 8 × 32 without multiplying: look up both numbers in a table of twos, add their counts,
and read across. Now you build the table and try it on any numbers you like.

Press **Run** on each grey box, top to bottom (or Shift + Enter). Then change the numbers and run it again.

```python
# Setup: in the browser version (JupyterLite) the sliders need one install; elsewhere this does nothing.
import sys
if "pyodide" in sys.modules:
    import piplite
    await piplite.install(["ipywidgets"])
```

```python
import math
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, FloatSlider
```

## 1 · Build the table

One column doubles; the other counts how many twos you multiplied. The counting number is the **logarithm, base 2**.

```python
for count in range(0, 21):
    print(f"{2 ** count:>12,}   {count:>2}")
```

## 2 · Multiply with it: look up, add, read across

```python
def multiply(a_count=3, b_count=5):
    a, b = 2 ** a_count, 2 ** b_count
    total = a_count + b_count
    print(f"{a:,} sits beside {a_count};  {b:,} sits beside {b_count}")
    print(f"add the counts: {a_count} + {b_count} = {total}")
    print(f"read across: beside {total} is {2 ** total:,}")
    print(f"check: {a:,} × {b:,} = {a * b:,}")

interact(multiply, a_count=IntSlider(value=3, min=0, max=15), b_count=IntSlider(value=5, min=0, max=15));
```

## 3 · The holes in the table

The powers of two leave bigger and bigger gaps. 3 and 7 are not in the table at all. Move the slider to see how
empty the number line is between the powers.

```python
def holes(top=64):
    powers = [2 ** k for k in range(0, 21) if 2 ** k <= top]
    fig, ax = plt.subplots(figsize=(10, 1.8))
    ax.plot([0, top], [0, 0], color="0.4")
    ax.scatter(powers, [0] * len(powers), color="tab:orange", zorder=3)
    for p in powers: ax.annotate(f"{p:,}", (p, 0), xytext=(0, 8), textcoords="offset points", ha="center", fontsize=8)
    ax.set_yticks([]); ax.set_xlim(0, top * 1.02); ax.set_title(f"powers of 2 up to {top:,}: {len(powers)} of {top:,} numbers")
    plt.show()

interact(holes, top=IntSlider(value=64, min=16, max=4096, step=16));
```

## 4 · Napier's answer: tiny steps

Napier multiplied by 0.9999999 instead of 2. Each step is just a subtraction: take away the number shifted seven
places. The steps are so small that the column has no holes.

```python
x = 10_000_000.0
for step in range(6):
    print(f"step {step}: {x:,.7f}")
    x = x - x / 10 ** 7

print("\nmultiply 0.9999999 by itself ten million times:", f"{0.9999999 ** 10_000_000:.7f}", " and 1/e =", f"{1 / math.e:.7f}")
```

## 5 · The three log rules, checked by the computer

```python
def rules(x=8.0, y=32.0, n=2):
    L = lambda v: math.log2(v)
    print(f"log2({x:g} × {y:g}) = {L(x * y):.4f}   and   log2 {x:g} + log2 {y:g} = {L(x) + L(y):.4f}")
    print(f"log2({x:g} ÷ {y:g}) = {L(x / y):.4f}   and   log2 {x:g} − log2 {y:g} = {L(x) - L(y):.4f}")
    print(f"log2({x:g}^{n})    = {L(x ** n):.4f}   and   {n} × log2 {x:g}     = {n * L(x):.4f}")

interact(rules, x=FloatSlider(value=8.0, min=0.5, max=100, step=0.5), y=FloatSlider(value=32.0, min=0.5, max=100, step=0.5),
         n=IntSlider(value=2, min=0, max=6));
```
