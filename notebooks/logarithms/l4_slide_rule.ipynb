# A slide rule you can slide

Put each number at the length of its log, and sliding one ruler along another adds lengths, so the numbers
multiply. Slide this one, then print the marks to make your own.

Press **Run** on each grey box, top to bottom (or Shift + Enter). Then change the numbers and run it again.

```python
# Setup: in the browser version (JupyterLite) the sliders need one install; elsewhere this does nothing.
import sys
if "pyodide" in sys.modules:
    import piplite
    await piplite.install(["ipywidgets"])
```

```python
import math
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
```

## 1 · Slide

```python
def scale(ax, y, offset, label):
    for v in [1 + k / 10 for k in range(10)] + [2 + k / 5 for k in range(15)] + [5 + k / 2 for k in range(11)]:
        x = offset + math.log10(v)
        major = abs(v - round(v)) < 1e-9
        ax.plot([x, x], [y, y + (0.25 if major else 0.12) * (1 if label == "slide" else -1)], color="k", lw=1.5 if major else 0.8)
        if major: ax.text(x, y + (0.35 if label == "slide" else -0.45), str(int(v)), ha="center", fontsize=9)

def slide_rule(a=2.0, b=3.0):
    fig, ax = plt.subplots(figsize=(11, 2.6))
    scale(ax, 0.1, math.log10(a), "slide")
    scale(ax, 0.0, 0.0, "stock")
    x = math.log10(a) + math.log10(b)
    ax.axvline(x, color="tab:orange", lw=2)
    ans = 10 ** x if x <= 1 else 10 ** (x - 1)
    ax.set_title(f"1 of the slide over {a:g}; under {b:g} read {ans:.3g}" + ("  (off the end: use the other index, ×10)" if x > 1 else ""))
    ax.set_xlim(-0.05, 2.05); ax.set_ylim(-0.7, 0.7); ax.axis("off"); plt.show()
    print(f"{a:g} × {b:g} = {a * b:g}")

interact(slide_rule, a=FloatSlider(value=2, min=1, max=10, step=0.1), b=FloatSlider(value=3, min=1, max=10, step=0.1));
```

## 2 · Make your own: marks for a 10 cm strip

```python
for n in range(1, 11):
    print(f"{n:>2} at {10 * math.log10(n):4.1f} cm")
```

## 3 · Log graph paper: a power law becomes a straight line

```python
def power_law(k=3.0, n=2.0):
    xs = [1 + 9 * i / 200 for i in range(201)]
    ys = [k * x ** n for x in xs]
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(10, 3.5))
    a1.plot(xs, ys); a1.set_title("ordinary axes")
    a2.loglog(xs, ys); a2.set_title(f"log-log axes: a straight line, slope {n:g}")
    for a in (a1, a2): a.grid(alpha=0.3)
    plt.show()

interact(power_law, k=FloatSlider(value=3, min=0.5, max=10, step=0.5), n=FloatSlider(value=2, min=0.5, max=4, step=0.5));
```
