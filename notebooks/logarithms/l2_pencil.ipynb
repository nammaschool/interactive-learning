## 1 · Near-misses give logs

Find a power of a number that almost matches a power of ten. If aⁿ ≈ 10ᵏ, then log a ≈ k ÷ n.

```python
def near_miss(a=2, n=10):
    v = a ** n
    k = round(math.log10(v))
    print(f"{a}^{n} = {v:,}   nearest power of ten: 10^{k} = {10 ** k:,}   ({100 * (v / 10 ** k - 1):+.1f} %)")
    print(f"so log {a} ≈ {k} ÷ {n} = {k / n:.4f}      true: {math.log10(a):.4f}")

interact(near_miss, a=IntSlider(value=2, min=2, max=9), n=IntSlider(value=10, min=1, max=30));
```

## 2 · Fill the table from a few primes

Every whole number is made by multiplying primes, and multiplying adds logs. With log 2, log 3, log 5 = 1 − log 2
and log 7, adding gives every number made from 2, 3, 5 and 7.

```python
L = {2: 0.30103, 3: 0.47712, 7: 0.84510}
L[5] = 1 - L[2]

def from_primes(n):
    total, m = 0.0, n
    for p in (2, 3, 5, 7):
        while m % p == 0: total += L[p]; m //= p
    return total if m == 1 else None

for n in range(2, 31):
    v = from_primes(n)
    print(f"log {n:>2} = {v:.5f}   (true {math.log10(n):.5f})" if v is not None else f"log {n:>2}: needs another prime")
```

## 3 · Briggs's recipe: root down, read off, double back

```python
def briggs(x=1.024, roots=10):
    r = x
    for _ in range(roots): r = math.sqrt(r)
    excess = r - 1
    near = excess * 0.4342944819        # near 1, log ≈ (amount above 1) × 0.434
    back = near * 2 ** roots              # each square root halved the log, so double back
    print(f"after {roots} square roots: {r:.12f}   (above 1 by {excess:.3e})")
    print(f"read off: {near:.6e};  double back {roots} times: log {x} ≈ {back:.8f}   true {math.log10(x):.8f}")

interact(briggs, roots=IntSlider(value=10, min=0, max=40));
```

Add 3 for the thousand and divide by 10, and you have log 2: (3 + log 1.024) ÷ 10.

```python
print(f"log 2 ≈ {(3 + math.log10(1.024)) / 10:.5f}")
```
