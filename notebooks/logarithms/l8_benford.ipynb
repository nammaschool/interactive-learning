# Test Benford's law yourself

About 30 % of numbers spread over many scales start with 1, because on a log scale 1 to 2 takes 30.1 % of every
decade. Test it on sequences, on growth, and on your own data.

Press **Run** on each grey box, top to bottom (or Shift + Enter). Then change the numbers and run it again.

```python
# Setup: in the browser version (JupyterLite) the sliders need one install; elsewhere this does nothing.
import sys
if "pyodide" in sys.modules:
    import piplite
    await piplite.install(["ipywidgets"])
```

```python
import math
from collections import Counter
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, FloatSlider

LAW = [math.log10(1 + 1 / d) for d in range(1, 10)]

def compare(numbers, title):
    c = Counter(int(str(abs(x)).lstrip("0.")[0]) for x in numbers if x)
    n = sum(c[d] for d in range(1, 10))
    fig, ax = plt.subplots(figsize=(8, 3))
    ax.bar(range(1, 10), [c[d] / n for d in range(1, 10)], alpha=0.6, label="your numbers")
    ax.plot(range(1, 10), LAW, "o-", color="k", label="Benford: log(1 + 1/d)")
    ax.set_xticks(range(1, 10)); ax.set_title(f"{title}: {c[1]} of {n} start with 1 ({100 * c[1] / n:.1f} %)"); ax.legend()
    plt.show()
```

## 1 · The law

```python
for d in range(1, 10):
    print(f"first digit {d}: {100 * LAW[d - 1]:4.1f} %")
```

## 2 · Powers of 2, and the Fibonacci numbers

```python
def sequences(n=1000):
    compare([2 ** k for k in range(1, n + 1)], f"first {n} powers of 2")
    fib, a, b = [], 1, 1
    for _ in range(n): fib.append(a); a, b = b, a + b
    compare(fib, f"first {n} Fibonacci numbers")

interact(sequences, n=IntSlider(value=1000, min=50, max=2000, step=50));
```

## 3 · Anything that grows by a fixed percentage

```python
def growth(rate=7.0, steps=500):
    x, values = 1.0, []
    for _ in range(steps): values.append(x); x *= 1 + rate / 100
    compare([int(v * 1000) for v in values], f"growing {rate:g} % a step")

interact(growth, rate=FloatSlider(value=7, min=0.5, max=50, step=0.5), steps=IntSlider(value=500, min=50, max=2000, step=50));
```
