# Compare quakes, sounds and acids

Decibels, pH and earthquake magnitude are all base-ten logs: one step up is one more multiplication, not one more
unit. Use these calculators to compare any two.

Press **Run** on each grey box, top to bottom (or Shift + Enter). Then change the numbers and run it again.

```python
# Setup: in the browser version (JupyterLite) the sliders need one install; elsewhere this does nothing.
import sys
if "pyodide" in sys.modules:
    import piplite
    await piplite.install(["ipywidgets"])
```

```python
import math
from ipywidgets import interact, FloatSlider
```

## 1 · Earthquakes: × 10 shaking and about × 31.6 energy per step

```python
def quakes(m1=7.6, m2=9.5):
    d = m2 - m1
    print(f"{d:+.1f} steps:  shaking × {10 ** d:,.1f}   energy × {10 ** (1.5 * d):,.0f}")

interact(quakes, m1=FloatSlider(value=7.6, min=2, max=9.5, step=0.1), m2=FloatSlider(value=9.5, min=2, max=9.5, step=0.1));
```

## 2 · Sound: + 10 dB = × 10 power

```python
def sound(db1=30.0, db2=60.0):
    print(f"{db2:g} dB carries {10 ** ((db2 - db1) / 10):,.1f} × the power of {db1:g} dB")
    print(f"doubling the power adds {10 * math.log10(2):.2f} dB")

interact(sound, db1=FloatSlider(value=30, min=0, max=140, step=1), db2=FloatSlider(value=60, min=0, max=140, step=1));
```

## 3 · pH: minus the log of the hydrogen-ion concentration

```python
def acidity(ph1=2.2, ph2=7.0):
    h1, h2 = 10 ** -ph1, 10 ** -ph2
    print(f"pH {ph1:g}: [H+] = {h1:.2e} mol/L     pH {ph2:g}: [H+] = {h2:.2e} mol/L")
    print(f"pH {ph1:g} has {h1 / h2:,.0f} × the hydrogen ions of pH {ph2:g}")

interact(acidity, ph1=FloatSlider(value=2.2, min=0, max=14, step=0.1), ph2=FloatSlider(value=7.0, min=0, max=14, step=0.1));
```
