# Use a log table

Move the decimal point and only the whole number of a base-ten log changes, so a table only needs the numbers from
1 to 10. Here you print a page of a four-figure table, read it, and multiply the way engineers did.

Press **Run** on each grey box, top to bottom (or Shift + Enter). Then change the numbers and run it again.

```python
# Setup: in the browser version (JupyterLite) the sliders need one install; elsewhere this does nothing.
import sys
if "pyodide" in sys.modules:
    import piplite
    await piplite.install(["ipywidgets"])
```

```python
import math
from ipywidgets import interact, IntSlider, FloatSlider
```

## 1 · Only the whole number changes

```python
for k in range(-2, 5):
    x = 2 * 10 ** k
    print(f"log {x:>10g} = {math.log10(x):8.4f}")
```

## 2 · Print a page of the table

Rows are the first two digits, columns the third. Each entry is a mantissa, four figures.

```python
def page(first_row=20):
    print("  N   " + "".join(f"{c:>6}" for c in range(10)))
    for r in range(first_row, min(first_row + 12, 100)):
        print(f" {r:>2}   " + "".join(f"{round(math.log10((10 * r + c) / 100), 4):.4f}"[2:].rjust(6) for c in range(10)))

interact(page, first_row=IntSlider(value=20, min=10, max=88));
```

## 3 · Characteristic + mantissa

```python
def split(x=234.0):
    char = math.floor(math.log10(x))
    mant = math.log10(x) - char
    print(f"log {x:g} = {char} + {mant:.4f} = {math.log10(x):.4f}")
    print(f"characteristic {char}: where the decimal point is")
    print(f"mantissa {mant:.4f}: from the table, for the digits {x / 10 ** char:.4g}")

interact(split, x=FloatSlider(value=234.0, min=0.01, max=99999, step=0.01));
```

## 4 · Multiply with the table: look up, add, split, antilog

```python
def multiply(a=23.4, b=567.0):
    la, lb = round(math.log10(a), 4), round(math.log10(b), 4)
    s = la + lb
    char, mant = math.floor(s), s - math.floor(s)
    anti = round(10 ** mant, 3)
    print(f"log {a:g} = {la:.4f}    log {b:g} = {lb:.4f}")
    print(f"add: {s:.4f}  →  characteristic {char}, mantissa {mant:.4f}")
    print(f"antilog {mant:.4f} = {anti}  →  {anti} × 10^{char} = {anti * 10 ** char:,.4g}")
    print(f"true answer: {a * b:,.4f}")

interact(multiply, a=FloatSlider(value=23.4, min=1, max=999, step=0.1), b=FloatSlider(value=567, min=1, max=999, step=1));
```
