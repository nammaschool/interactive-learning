# Walk 3 km, then 4 km. How far from home?

Walk three kilometres from your front door, turn, and walk four more. Seven is the answer that jumps out, but the
true answer is anything from one kilometre to seven: it depends on the turn. Distances don't add. Moves do, head to
tail, and a move is a size and a direction: a vector.

**How to use this notebook.** The grey boxes are cells. Click the first one and press **Shift + Enter** (or the ▶ Run
button) to run it. Run each cell in turn, from top to bottom. When a slider appears, drag it and watch. Change any
number and run the cell again: nothing can break. It works in your browser (no install) and in Google Colab.

```python
# Setup: in the browser version (JupyterLite) the sliders need one install; elsewhere this does nothing.
import sys
if "pyodide" in sys.modules:
    import piplite
    await piplite.install(["ipywidgets"])
```

```python
# The look: one style for every figure in this series of notebooks. Run this cell first.
import math
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

SLATE, SLATE2 = "#1f2b28", "#2a3a36"        # the slate ground
CHALK, DIM = "#ece6d6", "#93a097"           # chalk, and chalk smudge
CORAL = "#e3896b"                           # the one key thing in each figure
SUM, ALONG, ACROSS = "#a0c27c", "#6fb6bb", "#c9a0d3"   # the sum (resultant) · the along part · the across part

plt.rcParams.update({
    "figure.facecolor": SLATE, "axes.facecolor": SLATE, "savefig.facecolor": SLATE,
    "text.color": CHALK, "axes.labelcolor": CHALK, "axes.edgecolor": DIM, "axes.titlecolor": CHALK,
    "xtick.color": DIM, "ytick.color": DIM, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#3b4c47", "grid.linewidth": 0.8,
    "font.size": 13, "axes.titlesize": 15, "axes.labelsize": 13, "legend.fontsize": 12,
    "legend.facecolor": SLATE2, "legend.edgecolor": DIM, "legend.labelcolor": CHALK, "legend.framealpha": 0.9,
    "lines.linewidth": 2.4, "figure.dpi": 100,
})

def arrow(ax, p, q, col=CHALK, lw=2.8, ls="-"):
    """A vector arrow from point p to point q."""
    ax.annotate("", xy=q, xytext=p, arrowprops=dict(arrowstyle="-|>", color=col, lw=lw, ls=ls, mutation_scale=22))

def column(ax, vals, x, y, col=CHALK, label=""):
    """The arrow's numbers as a column in brackets, beside it: every arrow, two ways (axes coordinates)."""
    rows = "\n".join(f"{v:6.2f}" for v in vals)
    ax.text(x, y, rows, color=col, fontsize=16, family="monospace", va="top", ha="left", transform=ax.transAxes,
            bbox=dict(boxstyle="square,pad=0.45", fc=SLATE2, ec=DIM))
    if label: ax.text(x, y + 0.03, label, color=DIM, fontsize=12, va="bottom", ha="left", transform=ax.transAxes)

def walk(a, b, turn):
    """Walk a km east, then b km after turning `turn` degrees: the two moves and the move home, as numbers."""
    t = math.radians(turn)
    first = (a, 0.0)
    second = (b * math.cos(t), b * math.sin(t))
    home = (first[0] + second[0], first[1] + second[1])
    return first, second, home

def length(v):
    """The length of an arrow from its numbers (Pythagoras)."""
    return math.hypot(*v)
```

## 1 · Moves add head to tail

Three kilometres east is the column (3, 0): three across, none up. Four kilometres north is (0, 4). Put the tail of
the second at the head of the first, and the move home runs from where you started to where you finished. The numbers
add the same way, row by row.

```python
first, second, home = walk(3, 4, 90)
print("first walk  ", first)
print("second walk ", tuple(round(v, 6) for v in second))
print("home        ", tuple(round(v, 6) for v in home), " length", round(length(home), 6), "km")
first, second, home = walk(3, 4, 180)
print("turn right around: home", tuple(round(v, 6) for v in home), " length", round(length(home), 6), "km")
```

## 2 · Drag the turn

Run the cell, then drag **turn** from 0° (straight on) to 180° (right around). The first walk is chalk, the second
walk is chalk too, and the green dashed arrow is the way home. The columns beside the picture change with the arrows.

```python
def turn_it(turn=0):
    a, b = 3.0, 4.0
    first, second, home = walk(a, b, turn)
    fig, ax = plt.subplots(figsize=(11, 5.4))
    ax.plot([0], [0], "s", color=CHALK, ms=12)
    ax.text(0, -0.55, "door", color=DIM, ha="center", va="top")
    arrow(ax, (0, 0), first, CHALK)
    arrow(ax, first, home, CHALK)
    arrow(ax, (0, 0), home, SUM, lw=3.2, ls="--")
    if turn > 0:                                                       # the turn, marked at the corner
        th = np.radians(np.linspace(0, turn, 60))
        ax.plot(a + 0.8 * np.cos(th), 0.8 * np.sin(th), color=CORAL, lw=2.4)
        ax.text(a + 1.15 * math.cos(math.radians(turn / 2)), 1.15 * math.sin(math.radians(turn / 2)),
                f"{turn}°", color=CORAL, ha="center", va="center")
    ax.set_xlim(-2, 16); ax.set_ylim(-1.5, 5.2); ax.set_aspect("equal")
    ax.set_xlabel("east (km)"); ax.set_ylabel("north (km)")
    ax.set_title(f"turn {turn}°   ·   distance home {length(home):.2f} km", color=CORAL)
    column(ax, first, 0.50, 0.93, CHALK, "first walk")
    column(ax, second, 0.66, 0.93, CHALK, "second walk")
    column(ax, home, 0.83, 0.93, SUM, "home")
    ax.text(0.625, 0.84, "+", transform=ax.transAxes, fontsize=20, ha="center")
    ax.text(0.79, 0.84, "=", transform=ax.transAxes, fontsize=20, ha="center")
    plt.show()

interact(turn_it, turn=IntSlider(value=0, min=0, max=180, step=5, description="turn (°)"));
```

**Notice.** Straight on, home is 7 km away. At a right angle it is 5. Right around, it is 1. The way home is never
longer than the walk itself: |a + b| ≤ |a| + |b|.

## 3 · The formula

Textbooks print it in one line: |a + b|² = |a|² + |b|² + 2|a||b| cos θ, where θ is the turn between the two walks.
cos θ is 1 straight on, 0 at a right angle and −1 when you turn back: the straighter you keep walking, the further you
get from home. The columns and the formula always agree.

```python
print(f"{'turn':>6} {'formula':>22} {'from the columns':>18}")
for turn in (0, 60, 90, 120, 180):
    by_formula = math.sqrt(3**2 + 4**2 + 2 * 3 * 4 * math.cos(math.radians(turn)))
    print(f"{turn:>5}° {3**2 + 4**2 + 2 * 3 * 4 * math.cos(math.radians(turn)):>12.2f} → {by_formula:5.2f} km"
          f" {length(walk(3, 4, turn)[2]):>12.2f} km")
```

## 4 · Two walks of your own

Choose the two walks and the turn. Try the challenge from the video: walk 3 km, turn, and walk 3 km more. What turn
leaves you exactly 3 km from home?

```python
def your_walk(first_km=3.0, second_km=3.0, turn=90):
    first, second, home = walk(first_km, second_km, turn)
    fig, ax = plt.subplots(figsize=(9, 5))
    arrow(ax, (0, 0), first, CHALK)
    arrow(ax, first, home, CHALK)
    arrow(ax, (0, 0), home, SUM, lw=3.2, ls="--")
    ax.plot([0], [0], "s", color=CHALK, ms=10)
    lim = first_km + second_km + 1
    ax.set_xlim(-lim / 2, lim); ax.set_ylim(-0.5, lim * 0.62); ax.set_aspect("equal")
    ax.set_xlabel("east (km)"); ax.set_ylabel("north (km)")
    ax.set_title(f"home is {length(home):.2f} km away  (the walk was {first_km + second_km:.1f} km)", color=CORAL)
    column(ax, home, 0.80, 0.93, SUM, "home")
    plt.show()

interact(your_walk, first_km=FloatSlider(value=3, min=0.5, max=6, step=0.5, description="first (km)"),
         second_km=FloatSlider(value=3, min=0.5, max=6, step=0.5, description="second (km)"),
         turn=IntSlider(value=90, min=0, max=180, step=5, description="turn (°)"));
```
