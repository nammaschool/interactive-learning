```python
# The look: one style for every figure in this series of notebooks. Run this cell first.
import math
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

SLATE, SLATE2 = "#1f2b28", "#2a3a36"        # the slate ground
CHALK, DIM = "#ece6d6", "#93a097"           # chalk, and chalk smudge
CORAL = "#e3896b"                           # the one key thing in each figure
SUM, ALONG, ACROSS = "#a0c27c", "#6fb6bb", "#c9a0d3"   # the sum (resultant) · the along part · the across part

plt.rcParams.update({
    "figure.facecolor": SLATE, "axes.facecolor": SLATE, "savefig.facecolor": SLATE,
    "text.color": CHALK, "axes.labelcolor": CHALK, "axes.edgecolor": DIM, "axes.titlecolor": CHALK,
    "xtick.color": DIM, "ytick.color": DIM, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#3b4c47", "grid.linewidth": 0.8,
    "font.size": 13, "axes.titlesize": 15, "axes.labelsize": 13, "legend.fontsize": 12,
    "legend.facecolor": SLATE2, "legend.edgecolor": DIM, "legend.labelcolor": CHALK, "legend.framealpha": 0.9,
    "lines.linewidth": 2.4, "figure.dpi": 100,
})

def arrow(ax, p, q, col=CHALK, lw=2.6, label=None, at=None):
    """A vector arrow from point p to point q, with an optional label (beside its middle, or at a point)."""
    ax.annotate("", xy=q, xytext=p, arrowprops=dict(arrowstyle="-|>", color=col, lw=lw, mutation_scale=20))
    if label:
        x, y = at if at else ((p[0] + q[0]) / 2, (p[1] + q[1]) / 2)
        ax.text(x, y, label if at else "  " + label, color=col, fontsize=12, va="center", ha="center" if at else "left")

def column(ax, vals, unit, x, y, col=CHALK):
    """The arrow's numbers, as a column beside it: every arrow, two ways."""
    rows = "\n".join(f"{v:7.2f}" for v in vals)
    ax.text(x, y, rows, color=col, fontsize=15, family="monospace", va="top", ha="left",
            bbox=dict(boxstyle="square,pad=0.4", fc=SLATE2, ec=DIM))
    ax.text(x, y + 0.6, unit, color=DIM, fontsize=11, va="bottom", ha="left")

G = 9.8                                     # m/s², the pull of gravity used throughout

def size(*v):
    """The length of an arrow from its numbers (Pythagoras)."""
    return math.sqrt(sum(x * x for x in v))
```

## 1 · Rain and an umbrella

Follow one drop for one second: it falls 8 m and drifts 6 m with the wind. Its real path is the slanted side of a
right triangle, so it hits at √(6² + 8²) = 10 m/s, not 14. The umbrella's shaft points back along the rain, tilted
into the wind: tan θ = 6/8, about 36.9° from the vertical (53.1° from the ground).

Drag the wind and the rain. The arrow's column of numbers changes with it.

```python
def umbrella(wind=6.0, rain=8.0):
    v = (wind, -rain)
    fig, ax = plt.subplots(figsize=(8.6, 5.4))
    arrow(ax, (0, 0), (0, -rain), CHALK, label=f"falls {rain:g} m/s")
    arrow(ax, (0, -rain), (wind, -rain), CHALK, label=f"wind {wind:g} m/s", at=(wind / 2, -rain - 1.2))
    arrow(ax, (0, 0), v, SUM, lw=3.4)
    tilt = math.degrees(math.atan2(wind, rain))
    s = 3.0 / max(1e-9, size(*v))
    ax.plot([0, -v[0] * s], [0, -v[1] * s], color=CORAL, lw=3)          # the shaft, back along the rain
    ax.plot([-v[0] * s - 1.6 * v[1] * s / 3, -v[0] * s + 1.6 * v[1] * s / 3],
            [-v[1] * s + 1.6 * v[0] * s / 3, -v[1] * s - 1.6 * v[0] * s / 3], color=CORAL, lw=4)   # the canopy
    column(ax, v, "m/s", 11.5, 3.5, SUM)
    ax.set_xlim(-6, 16); ax.set_ylim(-14, 5); ax.set_aspect("equal")
    ax.set_xlabel("sideways (m/s)"); ax.set_ylabel("up (m/s)")
    ax.set_title(f"hits at {size(*v):.1f} m/s (not {wind + rain:g}) · tilt {tilt:.1f}° from the vertical")
    plt.show()

interact(umbrella, wind=FloatSlider(value=6, min=0, max=12, step=0.5, description="wind"),
         rain=FloatSlider(value=8, min=1, max=12, step=0.5, description="rain"));
```

## 2 · The method

1. **Draw the arrows**, each as long as its size and pointing its way.
2. **Split each arrow into two parts** at right angles, along two lines you choose. Your physics book calls the
   parts *components*.
3. **Add the parts**, line by line. The sum of the arrows is the *resultant*.
4. **Answer in words**, with units and a direction.

## 3 · Crossing a river

The river is 80 m wide and flows at 2 m/s; the boat moves at 4 m/s in still water. Head straight across and you
cross in 80 ÷ 4 = 20 s, but land 40 m downstream. Aim upstream by an angle θ: the upstream part 4 sin θ fights the
current, and the across part 4 cos θ carries you over.

Drag the heading upstream and watch the drift shrink. At 30° it reaches zero, and the crossing takes about 23.1 s.

```python
WIDTH, CURRENT, BOAT = 80, 2, 4

def river(heading=0):
    h = math.radians(heading)
    up, across = BOAT * math.sin(h), BOAT * math.cos(h)
    time = WIDTH / across
    drift = (CURRENT - up) * time
    if abs(drift) < 0.05: drift = 0.0
    fig, ax = plt.subplots(figsize=(8.6, 5.4))
    ax.axhspan(0, WIDTH, color=ALONG, alpha=0.10)
    ax.axhline(0, color=DIM); ax.axhline(WIDTH, color=DIM)
    for y in (12, 32, 52):
        arrow(ax, (-55, y), (-35, y), DIM, lw=1.6)
    ax.text(-58, -7, f"current {CURRENT} m/s →", color=DIM, fontsize=11, va="center")
    ax.plot([0, drift], [0, WIDTH], color=SUM, lw=3, label="the boat's track")
    arrow(ax, (0, 0), (-up * 5, across * 5), CHALK, label="heading")
    ax.plot([0], [WIDTH], "o", color=CORAL, ms=10, label="straight opposite")
    ax.set_xlim(-60, 60); ax.set_ylim(-10, 95); ax.set_aspect("equal")
    ax.set_xlabel("along the river (m), downstream →"); ax.set_ylabel("across (m)")
    ax.legend(loc="upper left", bbox_to_anchor=(0.0, 0.86))
    ax.set_title(f"heading {heading}° upstream · time {time:.1f} s · drift {drift:.1f} m")
    plt.show()
    print(f"upstream part 4 sin {heading}° = {up:.2f} m/s   across part 4 cos {heading}° = {across:.2f} m/s")

interact(river, heading=IntSlider(value=0, min=0, max=60, step=1, description="heading °"));
```

## 4 · Two ropes on a post

One rope pulls east with 8 N; the other pulls with 5 N at an angle θ to it. Split the 5 N into (5 cos θ, 5 sin θ),
add (8, 0), and put the parts back together with Pythagoras. At 60°: (10.5, 4.33), about 11.36 N. The textbook
formula R² = P² + Q² + 2PQ cos θ gives the same.

```python
def ropes(angle=60):
    t = math.radians(angle)
    q = (5 * math.cos(t), 5 * math.sin(t))
    r = (8 + q[0], q[1])
    fig, ax = plt.subplots(figsize=(8.6, 5.0))
    arrow(ax, (0, 0), (8, 0), CHALK, label="8 N")
    arrow(ax, (0, 0), q, CHALK, label="5 N")
    arrow(ax, (0, 0), r, SUM, lw=3.4)
    ax.plot([8, r[0]], [0, r[1]], color=DIM, ls="--", lw=1.2); ax.plot([q[0], r[0]], [q[1], r[1]], color=DIM, ls="--", lw=1.2)
    column(ax, r, "N", 14.5, 7, SUM)
    formula = math.sqrt(64 + 25 + 80 * math.cos(t))
    ax.set_xlim(-6, 19); ax.set_ylim(-2, 8); ax.set_aspect("equal")
    ax.set_title(f"θ = {angle}°: resultant {size(*r):.2f} N (the formula: {formula:.2f} N)")
    plt.show()

interact(ropes, angle=IntSlider(value=60, min=0, max=180, step=5, description="angle °"));
```

**Notice.** At 0° the ropes pull together: 13 N. At 180° they pull apart: 3 N. No angle gives less than 3 N, so
5 N and 8 N can never make 2 N.

## 5 · A block on a ramp

The weight mg points straight down. Split it into a part along the ramp, mg sin θ, which slides the block, and a
part into the ramp, mg cos θ. Tilt check: flat, nothing pulls along; a wall, the whole weight does. We use
g = 9.8 m/s².

```python
def ramp(angle=30, mass=5.0):
    t = math.radians(angle)
    w = mass * G
    along, into = w * math.sin(t), w * math.cos(t)
    fig, ax = plt.subplots(figsize=(8.6, 5.0))
    L = 10
    ax.fill([0, L * math.cos(t), L * math.cos(t)], [0, 0, L * math.sin(t)], color=SLATE2, ec=DIM)
    b = (0.55 * L * math.cos(t), 0.55 * L * math.sin(t))
    k = 3.5 / max(w, 1e-9)
    n = (-math.sin(t), math.cos(t))
    blk = [(b[0] + dx * math.cos(t) + dy * n[0], b[1] + dx * math.sin(t) + dy * n[1]) for dx, dy in ((-0.7, 0), (0.7, 0), (0.7, 1.2), (-0.7, 1.2))]
    ax.fill(*zip(*blk), color=SLATE, ec=CHALK, lw=1.6)
    arrow(ax, b, (b[0], b[1] - w * k), CHALK, label=f"mg = {w:.1f} N", at=(b[0] - 2.0, b[1] - w * k * 0.55))
    arrow(ax, b, (b[0] - along * k * math.cos(t), b[1] - along * k * math.sin(t)), ALONG)
    arrow(ax, b, (b[0] + into * k * math.sin(t), b[1] - into * k * math.cos(t)), ACROSS)
    ax.set_xlim(-2, 12); ax.set_ylim(-4.5, 9); ax.set_aspect("equal")
    ax.set_title(f"{angle}°: along {along:.1f} N (sin) · into {into:.1f} N (cos) · a = {G * math.sin(t):.2f} m/s²")
    plt.show()

interact(ramp, angle=IntSlider(value=30, min=0, max=85, step=5, description="angle °"),
         mass=FloatSlider(value=5, min=1, max=20, step=1, description="mass kg"));
```

## 6 · A ball in flight

Thrown at 20 m/s, 30° above the ground: across 20 cos 30° ≈ 17.32 m/s, up 20 sin 30° = 10 m/s. Gravity takes 9.8 off
the up part every second; the across part never changes (air resistance ignored). Drag the time.

```python
def flight(t=1.5):
    ux, uy = 20 * math.cos(math.radians(30)), 20 * math.sin(math.radians(30))
    ts = np.linspace(0, 2 * uy / G, 200)
    vy = uy - G * t
    x, y = ux * t, uy * t - G * t * t / 2
    fig, ax = plt.subplots(figsize=(8.6, 4.6))
    ax.plot(ux * ts, uy * ts - G * ts ** 2 / 2, color=DIM, lw=1.6)
    k = 0.4
    arrow(ax, (x, y), (x + ux * k, y), ALONG)
    arrow(ax, (x, y), (x, y + vy * k), ACROSS)
    arrow(ax, (x, y), (x + ux * k, y + vy * k), SUM, lw=3.2)
    ax.plot([x], [y], "o", color=CORAL, ms=10)
    ax.set_xlim(-2, 44); ax.set_ylim(-6, 10); ax.set_aspect("equal")
    word = "rising" if vy > 0 else "falling"
    ax.set_title(f"t = {t:.2f} s: velocity ({ux:.2f}, {vy:.2f}) m/s, {size(ux, vy):.1f} m/s, {word}")
    plt.show()

interact(flight, t=FloatSlider(value=1.5, min=0, max=2.0, step=0.05, description="t (s)"));
```

## 7 · Along or across? Work and torque

Work keeps the part of the force **along** the motion: a dot product. Torque keeps the part **across** the handle: a
cross product. Multiply the matching numbers and add, for the dot; the cross needs all three numbers.

```python
F = np.array([30 * math.cos(math.radians(40)), 30 * math.sin(math.radians(40))])
d = np.array([10.0, 0.0])
print(f"trolley: F = ({F[0]:.2f}, {F[1]:.2f}) N, d = (10, 0) m")
print(f"  F · d = {F[0]:.2f} × 10 + {F[1]:.2f} × 0 = {F @ d:.1f} J   (|F||d| cos 40° = {30 * 10 * math.cos(math.radians(40)):.1f} J)")
r, push = np.array([0.2, 0, 0]), np.array([30.0, 40.0, 0])
print(f"spanner: r × F = {np.round(np.cross(r, push), 6)} N m   F × r = {np.round(np.cross(push, r), 6)} N m")
print(f"  |τ| = rF sin θ = 0.2 × 50 × {40 / 50} = {0.2 * 50 * 40 / 50:.1f} N m")
```

## 8 · Check your answer

Type a problem's name, its numbers (separated by commas) and your answer, then press **check**. The cell works out
the answer from the numbers and tells you whether yours matches. The problems, and the numbers they take:

| problem | numbers | it checks |
|---|---|---|
| umbrella | rain, wind (m/s) | the speed the rain hits you (m/s) |
| river | width (m), current, boat (m/s) | the time to land straight opposite (s) |
| ropes | P, Q (N), angle (°) | the resultant (N) |
| ramp | mass (kg), angle (°) | the weight's part along the ramp (N) |
| flight | speed (m/s), angle (°), time (s) | the up part of the velocity (m/s) |
| work | force (N), angle (°), distance (m) | the work done (J) |
| torque | handle (m), push (N), angle (°) | the torque (N m) |
| crate | mass (kg), ramp length (m), rise (m) | the work pushing it up slowly (J) |

```python
from IPython.display import display
from ipywidgets import Text, Button, Output, HBox, VBox, Layout

def _rad(a): return math.radians(a)

def _river(w, cur, boat):
    if boat <= cur: return None, "the boat can't cancel the current, so it can't land straight opposite"
    across = boat * math.cos(math.asin(cur / boat))
    return w / across, "s"

PROBLEMS = {
    "umbrella": (2, lambda rain, wind: (size(rain, wind), "m/s")),
    "river":    (3, _river),
    "ropes":    (3, lambda p, q, a: (math.sqrt(p * p + q * q + 2 * p * q * math.cos(_rad(a))), "N")),
    "ramp":     (2, lambda m, a: (m * G * math.sin(_rad(a)), "N")),
    "flight":   (3, lambda u, a, t: (u * math.sin(_rad(a)) - G * t, "m/s")),
    "work":     (3, lambda f, a, s: (f * s * math.cos(_rad(a)), "J")),
    "torque":   (3, lambda r, f, a: (r * f * math.sin(_rad(a)), "N m")),
    "crate":    (3, lambda m, length, rise: (m * G * rise / length * length, "J")),
}

def check_your_answer(problem, numbers, yours):
    name = problem.strip().lower()
    if name not in PROBLEMS:
        print("I don't know that problem yet: try umbrella, river, ropes, ramp, flight, work, torque or crate."); return
    n, solve = PROBLEMS[name]
    try:
        nums = [float(s) for s in numbers.replace(";", ",").split(",") if s.strip()]
        mine = float(yours.replace("−", "-").split()[0]) if yours.strip() else None
    except ValueError:
        print("I can't read those numbers: write them like 8, 6 and your answer like 10."); return
    if len(nums) != n:
        print(f"{name} takes {n} numbers, separated by commas."); return
    answer, unit = solve(*nums)
    if answer is None:
        print(unit); return
    if mine is None:
        print("Type your answer, then press check."); return
    if abs(mine - answer) <= max(0.05, 0.01 * abs(answer)):
        print(f"✓ match: {answer:.2f} {unit}")
    else:
        print(f"✗ not a match. Draw the arrows, split them, add the parts, and try again.")

def checker(problem, numbers, yours):
    """Three boxes and a check button; the verdict prints under them."""
    problem_box = Text(value=problem, description="problem:", layout=Layout(width="230px"))
    numbers_box = Text(value=numbers, description="numbers:", layout=Layout(width="250px"))
    yours_box = Text(value=yours, description="yours:", layout=Layout(width="200px"))
    check_button, verdict = Button(description="check"), Output()
    def on_check(_=None):
        verdict.clear_output()
        with verdict: check_your_answer(problem_box.value, numbers_box.value, yours_box.value)
    check_button.on_click(on_check)
    display(VBox([HBox([problem_box, numbers_box, yours_box, check_button]), verdict]))
    on_check()
```

```python
checker("umbrella", "8, 6", "14")
```
