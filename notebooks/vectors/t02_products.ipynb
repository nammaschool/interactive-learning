# Dot product and cross product: along or across?

Three arrows a, b and c add to zero, and their lengths are 3, 4 and 5. What is a · b + b · c + c · a? The tempting
guess is 0. It's −25, and you can get it from the lengths alone, by dotting the sum with itself.

In this notebook you can multiply any two arrows both ways, redo every problem from the video, and check your own
answers.

```python
# The look: one style for every figure in this series of notebooks. Run this cell first.
import math
from fractions import Fraction
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

SLATE, SLATE2 = "#1f2b28", "#2a3a36"        # the slate ground
CHALK, DIM = "#ece6d6", "#93a097"           # chalk, and chalk smudge
CORAL = "#e3896b"                           # the one key thing in each figure
TEAL, SAGE, LILAC = "#6fb6bb", "#a0c27c", "#c9a0d3"   # along (dot) · the result · across (cross)

plt.rcParams.update({
    "figure.facecolor": SLATE, "axes.facecolor": SLATE, "savefig.facecolor": SLATE,
    "text.color": CHALK, "axes.labelcolor": CHALK, "axes.edgecolor": DIM, "axes.titlecolor": CHALK,
    "xtick.color": DIM, "ytick.color": DIM, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#3b4c47", "grid.linewidth": 0.8,
    "font.size": 13, "axes.titlesize": 15, "axes.labelsize": 13, "legend.fontsize": 12,
    "legend.facecolor": SLATE2, "legend.edgecolor": DIM, "legend.labelcolor": CHALK, "legend.framealpha": 0.9,
    "lines.linewidth": 2.4, "figure.dpi": 100,
})

def dot(a, b):
    """Multiply matching numbers and add."""
    return sum(x * y for x, y in zip(a, b))

def cross(a, b):
    """The determinant with i, j, k on top: note the minus sign on the middle term."""
    return (a[1] * b[2] - a[2] * b[1], -(a[0] * b[2] - a[2] * b[0]), a[0] * b[1] - a[1] * b[0])

def length(a):
    return math.sqrt(dot(a, a))

def show(v):
    """An arrow's numbers as a column would read them, with fractions kept exact."""
    return "(" + ", ".join(str(Fraction(x).limit_denominator(1000)) for x in v) + ")"

def column(ax, x, y, vals, color=CHALK, size=15):
    """An arrow's column of numbers, in square brackets, drawn at (x, y) on a figure."""
    rows = [str(Fraction(v).limit_denominator(100)) for v in vals]
    ax.text(x, y, "\n".join(rows), color=color, fontsize=size, va="center", ha="center", family="serif",
            linespacing=1.4, bbox=dict(boxstyle="square,pad=0.35", fc=SLATE, ec=DIM, lw=1.2))
```

## 1 · Every arrow, two ways

Drag the tips of two arrows from one point. Each arrow has its column of numbers beside it. The dot product
multiplies matching numbers and adds; the cross product (here, its one number straight out of the page) is the area of
the parallelogram the two arrows make.

```python
def two_arrows(a1=3.0, a2=1.0, b1=1.0, b2=3.0):
    a, b = (a1, a2), (b1, b2)
    fig, ax = plt.subplots(figsize=(8, 5.4))
    ax.fill([0, a1, a1 + b1, b1], [0, a2, a2 + b2, b2], color=LILAC, alpha=0.18)
    for v, col, name in ((a, CHALK, "a"), (b, CHALK, "b")):
        ax.annotate("", xy=v, xytext=(0, 0), arrowprops=dict(arrowstyle="-|>", color=col, lw=2.6, mutation_scale=22))
        ax.text(v[0] * 1.08, v[1] * 1.08, name, color=col, fontsize=16, family="serif")
    column(ax, 5.6, 3.0, a); column(ax, 6.6, 3.0, b)
    ax.text(5.6, 1.6, "a", ha="center", color=DIM, family="serif"); ax.text(6.6, 1.6, "b", ha="center", color=DIM, family="serif")
    ax.set_xlim(-4.5, 7.5); ax.set_ylim(-4.5, 4.5); ax.set_aspect("equal")
    ax.set_title(f"a · b = {a1:g}×{b1:g} + {a2:g}×{b2:g} = {dot(a, b):g}      area (a × b) = {a1 * b2 - a2 * b1:g}",
                 color=CHALK)
    plt.show()

interact(two_arrows, a1=FloatSlider(value=3, min=-4, max=4, step=0.5, description="a: across"),
         a2=FloatSlider(value=1, min=-4, max=4, step=0.5, description="a: up"),
         b1=FloatSlider(value=1, min=-4, max=4, step=0.5, description="b: across"),
         b2=FloatSlider(value=3, min=-4, max=4, step=0.5, description="b: up"));
```

## 2 · The trap: a + b + c = 0, with lengths 3, 4 and 5

Arrows that add to zero run head to tail and close a triangle. Sides 3, 4 and 5 make a right angle, so put
a = (3, 0, 0), b = (0, 4, 0) and c = (−3, −4, 0). Then dot the sum with itself, which needs only the lengths.

```python
a, b, c = (3, 0, 0), (0, 4, 0), (-3, -4, 0)
print("a + b + c =", tuple(x + y + z for x, y, z in zip(a, b, c)))
print("a · b =", dot(a, b), "  b · c =", dot(b, c), "  c · a =", dot(c, a))
print("total:", dot(a, b) + dot(b, c) + dot(c, a))
print("from the lengths alone: 0 = 9 + 16 + 25 + 2 × (sum), so the sum is", (0 - (9 + 16 + 25)) / 2)
```

## 3 · The method: along or across?

1. **Along or across?** Angles, projections, work and perpendicular arrows are along questions: the dot product.
   Areas, an arrow square to two others, and parallel arrows are across questions: the cross product.
2. **Write each arrow as its column of numbers.**
3. **Multiply**: matching numbers and add (dot), or the determinant (cross).
4. **Only lengths given?** Square the sum: dot it with itself.
5. **Check**: a perpendicular pair always gives zero.

## 4 · Along: angles, projections, work, perpendicular

The angle between (1, 2, 3) and (3, −1, 2); the part of a force F = (2, 3, −1) newtons along a track (1, 2, 2), and
its work over the track's 3 metres; and λ that makes (λ, 2, −1) perpendicular to (3, −1, 4).

```python
p, q = (1, 2, 3), (3, -1, 2)
cos_t = dot(p, q) / (length(p) * length(q))
print(f"angle: dot = {dot(p, q)}, lengths √{dot(p, p)} and √{dot(q, q)}, cos θ = {cos_t:.3f}, θ = {math.degrees(math.acos(cos_t)):.1f}°")
F, track = (2, 3, -1), (1, 2, 2)
print(f"projection of F on the track: {dot(F, track)} ÷ {length(track):g} = {dot(F, track) / length(track):g} N")
print(f"(divide by |F| instead and you'd get {dot(F, track) / length(F):.2f}: the wrong arrow)")
print(f"work over the track's {length(track):g} m: F · d = {dot(F, track)} J")
lam = Fraction(2 + 4, 3)
print(f"perpendicular: 3λ − 2 − 4 = 0 gives λ = {lam}; check: {dot((int(lam), 2, -1), (3, -1, 4))}")
```

## 5 · Only lengths: |a| = 3, |b| = 5, |a + b| = 7

Square the sum: 49 = 9 + 25 + 2 a · b. Then the difference: |a − b|² = 9 + 25 − 2 a · b.

```python
adb = (49 - 9 - 25) / 2
print(f"a · b = {adb}, cos θ = {adb / 15}, θ = {math.degrees(math.acos(adb / 15)):.0f}°")
print(f"|a − b|² = 9 + 25 − {2 * adb:g} = {9 + 25 - 2 * adb:g}, |a − b| = {math.sqrt(9 + 25 - 2 * adb):.2f}")
```

## 6 · Across: the cross product by determinant

a = (3, 2, 2), b = (0, 2, 1). The answer is square to both (both dot products are 0), its length is the
parallelogram's area, and a unit vector perpendicular to both is ± the answer divided by that length.

```python
a, b = (3, 2, 2), (0, 2, 1)
n = cross(a, b)
print("a × b =", n, "  check: n · a =", dot(n, a), ", n · b =", dot(n, b))
print("area of the parallelogram:", length(n))
print("unit vectors perpendicular to both: ±", show(tuple(Fraction(x, int(length(n))) for x in n)))
print("b × a =", cross(b, a), " (the arrow flips)")
print("parallel test: (2, −4, 6) × (−1, 2, −3) =", cross((2, -4, 6), (-1, 2, -3)))
```

## 7 · The case study: a solar panel on a roof

The panel's corners are P(2, 1, 0), Q(6, 1, 0) and R(2, 5, 3), in metres. Across finds which way the panel faces;
along finds how squarely the Sun (straight overhead, along k) meets it. Turn the view with the sliders.

```python
P, Q, R = np.array([2, 1, 0]), np.array([6, 1, 0]), np.array([2, 5, 3])
PQ, PR = Q - P, R - P
N = np.array(cross(PQ, PR))
unit = N / np.linalg.norm(N)
print("PQ =", tuple(PQ), " PR =", tuple(PR), " PQ × PR =", tuple(N), " length", np.linalg.norm(N))
print("area of the panel: half of", np.linalg.norm(N), "=", np.linalg.norm(N) / 2, "square metres")
print("unit normal: ±", show(unit), "   k · n =", show([unit[2]]))

def panel(turn=-60, tilt=25):
    fig = plt.figure(figsize=(7.4, 6))
    ax = fig.add_subplot(projection="3d")
    ax.set_facecolor(SLATE)
    for axis in (ax.xaxis, ax.yaxis, ax.zaxis):
        axis.set_pane_color((0.165, 0.227, 0.212, 1.0)); axis._axinfo["grid"]["color"] = "#3b4c47"
    ax.plot_trisurf([P[0], Q[0], R[0]], [P[1], Q[1], R[1]], [P[2], Q[2], R[2]], color=TEAL, alpha=0.35)
    for A, B in ((P, Q), (Q, R), (R, P)): ax.plot(*zip(A, B), color=CHALK, lw=2)
    ax.quiver(*P, *PQ, color=CHALK, lw=2, arrow_length_ratio=0.1)
    ax.quiver(*P, *PR, color=CHALK, lw=2, arrow_length_ratio=0.1)
    G = (P + Q + R) / 3
    ax.quiver(*G, *(3 * unit), color=LILAC, lw=2.6, arrow_length_ratio=0.15)
    ax.quiver(*G, 0, 0, 3, color=CORAL, lw=2, arrow_length_ratio=0.15)
    ax.text(*(G + 3.2 * unit), "n", color=LILAC); ax.text(*(G + [0, 0, 3.2]), "Sun", color=CORAL)
    ax.set_xlim(0, 7); ax.set_ylim(0, 7); ax.set_zlim(0, 6)
    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
    ax.view_init(elev=tilt, azim=turn)
    ax.set_title(f"area {np.linalg.norm(N) / 2:g} m²,  k · n = {unit[2]:.2f}")
    plt.show()

interact(panel, turn=IntSlider(value=-60, min=-180, max=180, step=10, description="turn"),
         tilt=IntSlider(value=25, min=0, max=80, step=5, description="tilt"));
```

## 8 · Check your answer

Two checkers. Type the numbers of an arrow separated by commas (fractions like `-2/7` are fine), then press
**check**. Run the first cell below once (it sets up the checkers), then the ones after it.

```python
from IPython.display import display
from ipywidgets import Button, Output, HBox, VBox, Layout, Text

def read(s):
    s = s.replace("−", "-").replace("(", "").replace(")", "")
    return tuple(Fraction(x.strip()) for x in s.split(",") if x.strip())

def fmt(x):
    return str(x).replace("-", "−")

def check_perpendicular(a, b, yours):
    """Is your answer square to both arrows? And is it a unit vector?"""
    try:
        a, b, y = read(a), read(b), read(yours)
    except Exception:
        print("I can't read that yet: type three numbers with commas, like 2, 1, 2"); return
    da, db = dot(y, a), dot(y, b)
    if da or db:
        bad = da if da else db
        print(f"✗ not perpendicular: dot = {fmt(bad)}")
        print(f"   yours · a = {fmt(da)},  yours · b = {fmt(db)}   (both must be 0)")
        print("   the right direction is a × b =", show(cross(a, b)), "(watch the middle sign)")
        return
    if dot(y, y) == 1: print("✓ a unit vector perpendicular to both (its negative works too)")
    else: print(f"✓ perpendicular to both; divide by its length, {length(y):.4g}, for a unit vector")

def check_dot(a, b, yours):
    """Is your number the dot product a · b?"""
    try:
        a, b, (y,) = read(a), read(b), read(yours)
    except Exception:
        print("I can't read that yet: arrows as three numbers with commas, your answer as one number"); return
    d = dot(a, b)
    angle = math.degrees(math.acos(max(-1, min(1, d / (length(a) * length(b)))))) if dot(a, a) and dot(b, b) else 0
    print(("✓ right: " if y == d else f"✗ not quite: yours is {fmt(y)}, but ") + f"a · b = {fmt(d)}  (the angle is {angle:.1f}°)")

def checker(fn, a, b, yours, label):
    """Three boxes and a check button; the verdict prints under them."""
    boxes = [Text(value=v, description=n, layout=Layout(width="230px")) for v, n in ((a, "a ="), (b, "b ="), (yours, label))]
    check_button, answer = Button(description="check"), Output()
    def on_check(_=None):
        answer.clear_output()
        with answer: fn(*(x.value for x in boxes))
    check_button.on_click(on_check)
    display(VBox([HBox(boxes + [check_button]), answer]))
    on_check()
```

**A vector perpendicular to both.** Type the two arrows and your answer (the direction is enough; the checker also
says whether it is a unit vector).

```python
checker(check_perpendicular, "3, 2, 2", "0, 2, 1", "-2/7, -3/7, 6/7", "yours:")
```

**A dot product.** Type the two arrows and your a · b.

```python
checker(check_dot, "1, 2, 3", "3, -1, 2", "7", "a · b:")
```

## 9 · Along and across, as the angle turns

Two arrows of lengths 3 and 4 from one point. Drag the angle from 0° to 180°: the dot product (the part along, in
teal) shrinks to zero at 90° and turns negative, while the cross product's area (the part across, in lilac) grows
to its largest at 90°.

```python
def turn(angle=0):
    th = math.radians(angle)
    a, b = np.array([3.0, 0]), 4 * np.array([math.cos(th), math.sin(th)])
    d, area = 12 * math.cos(th), 12 * math.sin(th)
    fig, (ax, bx) = plt.subplots(1, 2, figsize=(10, 4.6), gridspec_kw={"width_ratios": [1.3, 1]})
    ax.fill([0, a[0], a[0] + b[0], b[0]], [0, a[1], a[1] + b[1], b[1]], color=LILAC, alpha=0.25)
    ax.plot([0, b[0]], [0, 0], color=TEAL, lw=6, alpha=0.8, label="part of b along a")
    for v in (a, b): ax.annotate("", xy=v, xytext=(0, 0), arrowprops=dict(arrowstyle="-|>", color=CHALK, lw=2.6, mutation_scale=22))
    ax.set_xlim(-4.5, 7.5); ax.set_ylim(-1, 5); ax.set_aspect("equal"); ax.legend(loc="upper right")
    ax.set_title(f"θ = {angle}°")
    bx.bar(["a · b", "area |a × b|"], [d, area], color=[TEAL, LILAC])
    bx.axhline(0, color=DIM, lw=1); bx.set_ylim(-13, 13)
    bx.set_title(f"a · b = {d:+.2f}    area = {area:.2f}", color=CORAL if angle == 90 else CHALK)
    plt.show()

interact(turn, angle=IntSlider(value=0, min=0, max=180, step=10, description="angle °"));
```
