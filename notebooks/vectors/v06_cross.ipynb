# The bridge where multiplication broke: the cross product

On 16 October 1843, William Rowan Hamilton cut a multiplication into a Dublin bridge in which i times j is not j
times i. A spanner shows why arrows need one: only the part of a push **across** the handle turns the bolt, and the
turn has a direction, along the bolt. That is the **cross product**.

```python
# The look: one style for every figure in these notebooks. Run this cell first.
import math
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

SLATE, SLATE2 = "#1f2b28", "#2a3a36"        # the slate ground
CHALK, DIM = "#ece6d6", "#93a097"           # chalk, and chalk smudge
CORAL = "#e3896b"                           # the one key thing in each figure
TEAL, SAGE, LILAC = "#6fb6bb", "#a0c27c", "#c9a0d3"   # the "along" part · a result · the "across" part

plt.rcParams.update({
    "figure.facecolor": SLATE, "axes.facecolor": SLATE, "savefig.facecolor": SLATE,
    "text.color": CHALK, "axes.labelcolor": CHALK, "axes.edgecolor": DIM, "axes.titlecolor": CHALK,
    "xtick.color": DIM, "ytick.color": DIM, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#3b4c47", "grid.linewidth": 0.8,
    "font.size": 13, "axes.titlesize": 15, "axes.labelsize": 13, "legend.fontsize": 12,
    "legend.facecolor": SLATE2, "legend.edgecolor": DIM, "legend.labelcolor": CHALK, "legend.framealpha": 0.9,
    "lines.linewidth": 2.4, "figure.dpi": 100,
})

def dot(a, b):
    """Multiply the matching numbers, then add."""
    return sum(p * q for p, q in zip(a, b))

def cross(a, b):
    """The cross product from the grid recipe: cover a column, cross-multiply the four numbers left, subtract.
    The signs go plus, minus, plus."""
    return (a[1] * b[2] - a[2] * b[1],
            -(a[0] * b[2] - a[2] * b[0]),
            a[0] * b[1] - a[1] * b[0])

def mag(a):
    return math.sqrt(dot(a, a))

def arrow(ax, tail, tip, color=CHALK, lw=3, label=None):
    ax.annotate("", xy=tip, xytext=tail, arrowprops=dict(arrowstyle="-|>", color=color, lw=lw, mutation_scale=20))
    if label: ax.annotate(label, tip, xytext=(8, 6), textcoords="offset points", color=color)

def column(ax, nums, x, y, color=CHALK, size=17):
    """An arrow's numbers as a column in brackets, at (x, y) in axes fractions."""
    body = "\n".join(f"{v:.2f}".rstrip("0").rstrip(".").replace("-", "−") for v in nums)
    ax.text(x, y, body, transform=ax.transAxes, color=color, fontsize=size, family="serif", va="center", ha="center",
            linespacing=1.5, bbox=dict(boxstyle="square,pad=0.5", fc=SLATE2, ec=DIM))
```

## 1 · Every arrow, two ways

An arrow on the board is also its numbers. Drag the sliders and watch the column change with the arrow.

```python
def two_ways(angle=30, length=4.0):
    a = (length * math.cos(math.radians(angle)), length * math.sin(math.radians(angle)))
    fig, ax = plt.subplots(figsize=(8, 5))
    arrow(ax, (0, 0), a, CORAL, 3.5)
    ax.set_xlim(-5, 5); ax.set_ylim(-5, 5); ax.set_aspect("equal")
    ax.set_title(f"length {length:.1f} at {angle}°")
    column(ax, a, 1.08, 0.5, CORAL)
    plt.show()

interact(two_ways, angle=IntSlider(value=30, min=-180, max=180, step=1, description="angle (°)"),
         length=FloatSlider(value=4.0, min=0.5, max=5.0, step=0.5, description="length"));
```

## 2 · Turn the spanner

A spanner 0.25 m long, seen from above, and a push of 40 newtons at its end. Drag the slider to change the angle of
the push from the handle. The push splits into a teal part **along** the handle and a lilac part **across** it. Only
the across part turns the bolt: the turn is the handle's length times the across part.

```python
def spanner(push_angle=0):
    r, F = 0.25, 40
    th = math.radians(push_angle)
    push = (F * math.cos(th), F * math.sin(th))
    turn = cross((r, 0, 0), (push[0], push[1], 0))[2]
    fig, (ax, bx) = plt.subplots(1, 2, figsize=(11.5, 4.8), gridspec_kw={"width_ratios": [1.5, 1]})
    ax.plot([0, r], [0, 0], color=DIM, lw=14, solid_capstyle="round")
    ax.add_patch(plt.Polygon([(0.03 * math.cos(k * math.pi / 3), 0.03 * math.sin(k * math.pi / 3)) for k in range(6)],
                             fc=SLATE2, ec=CHALK, lw=2))
    s = 0.004                                                       # newtons → metres on the drawing
    arrow(ax, (r, 0), (r + s * push[0], s * push[1]), CORAL, 3, "push")
    ax.plot([r, r + s * push[0]], [0, 0], color=TEAL, lw=6, alpha=0.8)
    ax.plot([r + s * push[0], r + s * push[0]], [0, s * push[1]], color=LILAC, lw=6, alpha=0.8)
    ax.set_xlim(-0.08, 0.5); ax.set_ylim(-0.2, 0.2); ax.set_aspect("equal"); ax.grid(False)
    ax.set_title(f"push at {push_angle}° to the handle: along {push[0]:.1f} N, across {push[1]:.1f} N")
    column(ax, (round(push[0], 1), round(push[1], 1), 0), 1.06, 0.5, CORAL)
    angs = np.linspace(0, 180, 181)
    bx.plot(angs, r * F * np.sin(np.radians(angs)), color=SAGE)
    bx.plot([push_angle], [turn], "o", color=CORAL, ms=10)
    bx.set_xlabel("angle of the push (°)"); bx.set_ylabel("turn (newton metres)"); bx.set_ylim(-0.5, 11)
    bx.set_title(f"turn = 0.25 × {push[1]:.1f} = {turn:.2f} N m")
    plt.tight_layout(); plt.show()

interact(spanner, push_angle=IntSlider(value=0, min=0, max=180, step=10, description="push angle (°)"));
```

**Notice.** Pushing along the handle (0°) turns nothing. At 30° the across part is 20 N and the turn is 5 newton
metres. Straight across (90°) gives the biggest turn, 10 newton metres.

## 3 · A cross product in 3D, with the right hand

a = (3, 0, 0) and b = (1, 2, 0) make a parallelogram of base 3 and height 2. Their cross product stands straight up
from it, as long as its area. Curl the fingers of your right hand from a toward b: your thumb points along a × b.
Drag the slider to walk round it.

```python
from mpl_toolkits.mplot3d import Axes3D   # noqa: F401  (registers the 3D axes)

def slate_3d(ax):
    """3D axes on the slate: dark panes, faint grid lines."""
    ax.set_facecolor(SLATE)
    for axis in (ax.xaxis, ax.yaxis, ax.zaxis):
        axis.set_pane_color((0.165, 0.227, 0.212, 1.0))
        try: axis._axinfo["grid"].update(color="#3b4c47", linewidth=0.8)
        except Exception: pass

def in_3d(view=-60):
    a, b = (3, 0, 0), (1, 2, 0)
    n = cross(a, b)
    fig = plt.figure(figsize=(7.5, 6)); ax = fig.add_subplot(projection="3d")
    slate_3d(ax)
    P = np.array([(0, 0, 0), a, (a[0] + b[0], a[1] + b[1], 0), b])
    ax.plot_trisurf(P[:, 0], P[:, 1], P[:, 2], color=SAGE, alpha=0.35, triangles=[[0, 1, 2], [0, 2, 3]])
    ax.quiver(0, 0, 0, *a, color=CHALK, lw=3, arrow_length_ratio=0.1)
    ax.quiver(0, 0, 0, *b, color=CHALK, lw=3, arrow_length_ratio=0.15)
    ax.quiver(0, 0, 0, *n, color=LILAC, lw=3, arrow_length_ratio=0.08)
    t = np.linspace(0, math.atan2(2, 1) * 0.9, 40)                 # the fingers curl from a toward b
    ax.plot(1.2 * np.cos(t), 1.2 * np.sin(t), 0.6 + 0 * t, color=CORAL, lw=2.5)
    ax.text(1.3, 0.9, 0.9, "fingers: a → b", color=CORAL); ax.text(0.2, 0, n[2] + 0.4, "thumb: a × b", color=LILAC)
    ax.set_xlim(0, 4); ax.set_ylim(0, 4); ax.set_zlim(0, 6.5); ax.view_init(elev=24, azim=view)
    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
    ax.set_title(f"a × b = {n}, length {mag(n):.0f} = the area, base 3 × height 2")
    plt.show()

interact(in_3d, view=IntSlider(value=-60, min=-180, max=180, step=10, description="view (°)"));
```

## 4 · The grid recipe, and the check

The triangle with corners A(1, 0, 0), B(0, 2, 0), C(0, 0, 2). Its area is half the length of AB × AC. Check any
cross product by dotting it with each arrow: both must give zero.

```python
A, B, C = (1, 0, 0), (0, 2, 0), (0, 0, 2)
AB = tuple(q - p for p, q in zip(A, B)); AC = tuple(q - p for p, q in zip(A, C))
N = cross(AB, AC)
print("AB =", AB, "  AC =", AC)
print("AB × AC =", N, "  length √", dot(N, N), "≈", round(mag(N), 2))
print("area of the triangle = half of it ≈", round(mag(N) / 2, 2), "(root six)")
print("check: N · AB =", dot(N, AB), "  N · AC =", dot(N, AC))
slip = (N[0], -N[1], N[2])
print("the slip with + in the middle:", slip, "  slip · AB =", dot(slip, AB), "(not zero, so it is wrong)")
```
