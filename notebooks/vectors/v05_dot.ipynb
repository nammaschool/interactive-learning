# Why is it winter when Earth is closest to the Sun? The dot product

On 3 January Earth is as close to the Sun as it gets all year, and January sunlight is about 7% stronger than July's.
Yet January is winter in the whole northern half of the world. The reason is the angle of the light: only the part of
the sunlight that points along the ground's "up" arrow heats the ground. That "how much lies along" is the **dot
product**.

```python
# The look: one style for every figure in these notebooks. Run this cell first.
import math
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

SLATE, SLATE2 = "#1f2b28", "#2a3a36"        # the slate ground
CHALK, DIM = "#ece6d6", "#93a097"           # chalk, and chalk smudge
CORAL = "#e3896b"                           # the one key thing in each figure
TEAL, SAGE, LILAC = "#6fb6bb", "#a0c27c", "#c9a0d3"   # the "along" part · a result · the "across" part

plt.rcParams.update({
    "figure.facecolor": SLATE, "axes.facecolor": SLATE, "savefig.facecolor": SLATE,
    "text.color": CHALK, "axes.labelcolor": CHALK, "axes.edgecolor": DIM, "axes.titlecolor": CHALK,
    "xtick.color": DIM, "ytick.color": DIM, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#3b4c47", "grid.linewidth": 0.8,
    "font.size": 13, "axes.titlesize": 15, "axes.labelsize": 13, "legend.fontsize": 12,
    "legend.facecolor": SLATE2, "legend.edgecolor": DIM, "legend.labelcolor": CHALK, "legend.framealpha": 0.9,
    "lines.linewidth": 2.4, "figure.dpi": 100,
})

def dot(a, b):
    """The dot product from the numbers: multiply the matching numbers, then add."""
    return sum(p * q for p, q in zip(a, b))

def mag(a):
    """The magnitude (length) of an arrow: the square root of the arrow dotted with itself."""
    return math.sqrt(dot(a, a))

def arrow(ax, tail, tip, color=CHALK, lw=3, label=None):
    """An arrow from tail to tip, with an optional label at its tip."""
    ax.annotate("", xy=tip, xytext=tail, arrowprops=dict(arrowstyle="-|>", color=color, lw=lw, mutation_scale=20))
    if label: ax.annotate(label, tip, xytext=(8, 6), textcoords="offset points", color=color)

def column(ax, nums, x, y, color=CHALK, size=17):
    """An arrow's numbers as a column in brackets, at (x, y) in axes fractions."""
    body = "\n".join(f"{v:.2f}".rstrip("0").rstrip(".").replace("-", "−") for v in nums)
    ax.text(x, y, body, transform=ax.transAxes, color=color, fontsize=size, family="serif", va="center", ha="center",
            linespacing=1.5, bbox=dict(boxstyle="square,pad=0.5", fc=SLATE2, ec=DIM))
```

## 1 · Every arrow, two ways

An arrow on the board can also be written as its numbers: how far across, and how far up. Drag the sliders to turn
and stretch the arrow, and watch its column of numbers change with it.

```python
def two_ways(angle=37, length=5.0):
    a = (length * math.cos(math.radians(angle)), length * math.sin(math.radians(angle)))
    fig, ax = plt.subplots(figsize=(8, 5))
    arrow(ax, (0, 0), a, CORAL, 3.5)
    ax.plot([0, a[0]], [0, 0], color=TEAL, lw=2, ls="--"); ax.plot([a[0], a[0]], [0, a[1]], color=LILAC, lw=2, ls="--")
    ax.set_xlim(-6, 6); ax.set_ylim(-6, 6); ax.set_aspect("equal")
    ax.set_title(f"length {length:.1f} at {angle}°: across {a[0]:.2f}, up {a[1]:.2f}")
    column(ax, a, 1.08, 0.5, CORAL)
    plt.show()

interact(two_ways, angle=IntSlider(value=37, min=-180, max=180, step=1, description="angle (°)"),
         length=FloatSlider(value=5.0, min=0.5, max=6.0, step=0.5, description="length"));
```

**Notice.** At 37° and length 5, the numbers are about 4 across and 3 up. The picture and the column always say the
same thing.

## 2 · Tilt the field under the Sun

The Sun is low: 15° above the horizon, London's noon on the shortest day. The coral arrow is the sunlight, one unit
long, pointing toward the Sun. The chalk arrow is the field's **up arrow**, one unit long, straight out of the ground.
Drag the slider to tilt the field (a hillside facing the Sun). The teal part of the sunlight lies **along** the up
arrow: only that part heats the ground.

The dot product is worked out two ways side by side: from the angle between the arrows, and from their numbers.

```python
def tilt(field_tilt=0, sun_height=15):
    s = (math.cos(math.radians(sun_height)), math.sin(math.radians(sun_height)))       # toward the Sun
    n = (math.sin(math.radians(field_tilt)), math.cos(math.radians(field_tilt)))      # the field's up arrow
    theta = math.degrees(math.acos(max(-1.0, min(1.0, dot(s, n)))))
    by_angle = mag(s) * mag(n) * math.cos(math.radians(theta))
    by_numbers = s[0] * n[0] + s[1] * n[1]
    fig, (ax, bx) = plt.subplots(1, 2, figsize=(11.5, 5), gridspec_kw={"width_ratios": [1.4, 1]})
    g = np.array([math.cos(math.radians(field_tilt)), -math.sin(math.radians(field_tilt))])  # along the field
    ax.fill([-1.6 * g[0], 1.6 * g[0], 1.6 * g[0], -1.6 * g[0]], [-1.6 * g[1], 1.6 * g[1], -1.8, -1.8], color=SLATE2)
    ax.plot([-1.6 * g[0], 1.6 * g[0]], [-1.6 * g[1], 1.6 * g[1]], color=DIM, lw=3)
    arrow(ax, (0, 0), n, CHALK, 3, "up arrow")
    arrow(ax, (0, 0), s, CORAL, 3, "sunlight")
    along = (by_numbers * n[0], by_numbers * n[1])
    ax.plot([0, along[0]], [0, along[1]], color=TEAL, lw=7, alpha=0.8, solid_capstyle="butt")
    ax.plot([s[0], along[0]], [s[1], along[1]], color=DIM, lw=1.2, ls=":")
    ax.set_xlim(-1.7, 1.7); ax.set_ylim(-1.2, 1.4); ax.set_aspect("equal"); ax.grid(False)
    ax.set_title(f"the Sun {sun_height}° up · the field tilted {field_tilt}°")
    bx.axis("off")
    rows = [("the angle way", f"|s| |n| cos {theta:.1f}°", f"= {by_angle:.3f}"),
            ("the numbers way", f"{s[0]:.3f} × {n[0]:.3f} + {s[1]:.3f} × {n[1]:.3f}".replace("-", "−"), f"= {by_numbers:.3f}")]
    for k, (head, sum_, val) in enumerate(rows):
        y = 0.82 - 0.38 * k
        bx.text(0.0, y, head, color=DIM, fontsize=13, transform=bx.transAxes)
        bx.text(0.0, y - 0.1, sum_, color=CHALK, fontsize=14, family="serif", transform=bx.transAxes)
        bx.text(0.0, y - 0.2, val, color=TEAL, fontsize=22, family="serif", transform=bx.transAxes)
    bx.text(0.0, 0.0, f"the ground gets {by_numbers:.0%} of the full beam", color=CORAL, fontsize=14,
            transform=bx.transAxes)
    plt.tight_layout(); plt.show()

interact(tilt, field_tilt=IntSlider(value=0, min=0, max=90, step=15, description="field tilt (°)"),
         sun_height=IntSlider(value=15, min=0, max=90, step=1, description="Sun height (°)"));
```

**Notice.** Flat, the field gets only 0.26 of the beam. Tilt it 75° toward the Sun and the up arrow points straight
at the Sun: the dot product is 1, the whole beam. The two ways always agree. Now set the Sun to 62°, London's June
noon: flat ground gets 0.88, about 3.4 times the winter share.

```python
for place, lat in (("London", 51.51), ("New Delhi", 28.61)):
    for month, decl in (("December", -23.44), ("June", 23.44)):
        h = 90 - abs(lat - decl)
        print(f"{place:10} {month:9} noon Sun {h:5.1f}°   up part {math.sin(math.radians(h)):.2f}   "
              f"a beam 1 m across lands on {1 / math.sin(math.radians(h)):.2f} m of ground")
```

## 3 · Work done by a force

A suitcase pulled with 40 newtons, the handle at 60° to the floor, for 10 metres. Only the part of the pull along
the floor does work.

```python
pull = (40 * math.cos(math.radians(60)), 40 * math.sin(math.radians(60)))
walk = (10, 0)
print("pull as numbers:", tuple(round(v, 1) for v in pull))
print("the angle way:   40 × 10 × cos 60° =", round(40 * 10 * math.cos(math.radians(60)), 6), "joules")
print("the numbers way:", f"{pull[0]:.1f} × 10 + {pull[1]:.1f} × 0 =", round(dot(pull, walk), 6), "joules")
```

## 4 · An angle nobody tells you, in 3D

Two diagonals across two faces of a cube, from the same corner: (1, 1, 0) and (0, 1, 1). Drag the slider to walk
round the cube. The dot product finds the angle between them.

```python
from mpl_toolkits.mplot3d import Axes3D   # noqa: F401  (registers the 3D axes)

def slate_3d(ax):
    """3D axes on the slate: dark panes, faint grid lines."""
    ax.set_facecolor(SLATE)
    for axis in (ax.xaxis, ax.yaxis, ax.zaxis):
        axis.set_pane_color((0.165, 0.227, 0.212, 1.0))
        try: axis._axinfo["grid"].update(color="#3b4c47", linewidth=0.8)
        except Exception: pass

def cube(view=-60):
    a, b = (1, 1, 0), (0, 1, 1)
    cos_t = dot(a, b) / (mag(a) * mag(b))
    fig = plt.figure(figsize=(7, 6)); ax = fig.add_subplot(projection="3d")
    slate_3d(ax)
    for s in (0, 1):
        for t in (0, 1):
            ax.plot([0, 1], [s, s], [t, t], color=DIM, lw=1); ax.plot([s, s], [0, 1], [t, t], color=DIM, lw=1)
            ax.plot([s, s], [t, t], [0, 1], color=DIM, lw=1)
    ax.quiver(0, 0, 0, *a, color=CORAL, lw=3, arrow_length_ratio=0.12)
    ax.quiver(0, 0, 0, *b, color=TEAL, lw=3, arrow_length_ratio=0.12)
    ax.plot([a[0], b[0]], [a[1], b[1]], [a[2], b[2]], color=SAGE, lw=2, ls="--")
    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
    ax.set_xlim(0, 1); ax.set_ylim(0, 1); ax.set_zlim(0, 1); ax.view_init(elev=22, azim=view)
    ax.set_title(f"a · b = {dot(a, b)}, |a| = |b| = √2, cos θ = {cos_t:.2f}, θ = {math.degrees(math.acos(cos_t)):.0f}°")
    plt.show()

interact(cube, view=IntSlider(value=-60, min=-180, max=180, step=10, description="view (°)"));
```

**Notice.** The dashed line joining the tips is a third face diagonal, so the triangle has three equal sides, and
every corner is 60°.

## 5 · The exam question: angle and projection

a = 2i + 3j + 6k and b = i + 2j + 2k. The projection of a on b is how much of a lies along b: a · b ÷ |b|. Divide by
the magnitude of b, the direction you measure along, not by the magnitude of a.

```python
a, b = (2, 3, 6), (1, 2, 2)
print("a · b =", dot(a, b), "  |a| =", mag(a), "  |b| =", mag(b))
print("cos θ = 20/21 =", round(dot(a, b) / (mag(a) * mag(b)), 4), "  θ ≈", round(math.degrees(math.acos(20 / 21)), 1), "°")
print("projection of a on b = a · b ÷ |b| =", round(dot(a, b) / mag(b), 2))
print("the slip, a · b ÷ |a| =", round(dot(a, b) / mag(a), 2), " (far too short: most of a's 7 units lie along b)")
```
