# Every derivative in the table, from one picture

Slide a tangent along a curve and mark how steep it is, point by point: the marks join into a second curve, the
curve's **slope shadow**. Every line of the table of derivatives is the answer to one question: what shadow does
this curve cast? Constant and straight lines, powers, eˣ and aˣ, ln x and log base a, sine, cosine and tangent, and
the inverse functions.

**How to use this notebook.** The grey boxes are cells. Click the first one and press **Shift + Enter** (or the
Run button) to run it, then run each cell in turn, from top to bottom. When the buttons and the slider appear,
pick a curve and drag the slider: the tangent slides along the curve and its shadow draws underneath. The last
cell hides one column of the table: try to fill it in. Nothing can break: if something looks odd, run everything
again from the top. It works in your browser (no install) and in Google Colab.

```python
# Setup: in the browser version (JupyterLite) the sliders need one install; elsewhere this does nothing.
import sys
if "pyodide" in sys.modules:
    import piplite
    await piplite.install(["ipywidgets"])
```

```python
# The look: one style for every figure in this series of notebooks. Run this cell first.
import math
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, ToggleButtons

SLATE, SLATE2 = "#1f2b28", "#2a3a36"        # the slate ground
CHALK, DIM = "#ece6d6", "#93a097"           # chalk, and chalk smudge
CORAL = "#e3896b"                           # the one key thing in each figure
TEAL, SAGE, LILAC = "#6fb6bb", "#a0c27c", "#c9a0d3"   # teal: the curve · sage: its slope · lilac: areas

plt.rcParams.update({
    "figure.facecolor": SLATE, "axes.facecolor": SLATE, "savefig.facecolor": SLATE,
    "text.color": CHALK, "axes.labelcolor": CHALK, "axes.edgecolor": DIM, "axes.titlecolor": CHALK,
    "xtick.color": DIM, "ytick.color": DIM, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#3b4c47", "grid.linewidth": 0.8,
    "font.size": 13, "axes.titlesize": 15, "axes.labelsize": 13, "legend.fontsize": 12,
    "legend.facecolor": SLATE2, "legend.edgecolor": DIM, "legend.labelcolor": CHALK, "legend.framealpha": 0.9,
    "lines.linewidth": 2.4, "figure.dpi": 100,
    "axes.prop_cycle": plt.cycler(color=[CHALK, TEAL, SAGE, LILAC, CORAL]),
})

def slope(f, x, h=1e-6):
    """The slope of f at x, measured over a tiny window: (f(x + h) − f(x − h)) ÷ 2h."""
    return (f(x + h) - f(x - h)) / (2 * h)

# Each curve: its function, the stretch of x we draw, the y range of the curve and of its shadow, and its shadow's name.
CURVES = {
    "4":       (lambda x: 4 + 0 * x,        (-3, 3),      (-1, 6),     (-1, 4),     "0"),
    "3x + 2":  (lambda x: 3 * x + 2,        (-3, 3),      (-8, 12),    (-1, 4),     "3"),
    "x²":      (lambda x: x ** 2,           (-3, 3),      (-1, 9.5),   (-6.5, 6.5), "2x"),
    "x³":      (lambda x: x ** 3,           (-2.2, 2.2),  (-11, 11),   (-1, 15),    "3x²"),
    "√x":      (np.sqrt,                    (0.05, 6),    (0, 2.6),    (0, 2.4),    "1 ÷ (2√x)"),
    "eˣ":      (np.exp,                     (-3, 2),      (0, 7.5),    (0, 7.5),    "eˣ, itself"),
    "2ˣ":      (lambda x: 2.0 ** x,         (-3, 2.6),    (0, 6.5),    (0, 6.5),    "2ˣ × ln 2 = 0.693 × 2ˣ"),
    "ln x":    (np.log,                     (0.12, 8),    (-2.2, 2.2), (0, 4),      "1 ÷ x"),
    "log₁₀ x": (np.log10,                   (0.12, 8),    (-1, 1),     (0, 2),      "1 ÷ (x ln 10)"),
    "sin x":   (np.sin,                     (0, 2 * np.pi), (-1.3, 1.3), (-1.3, 1.3), "cos x"),
    "cos x":   (np.cos,                     (0, 2 * np.pi), (-1.3, 1.3), (-1.3, 1.3), "−sin x"),
    "tan x":   (np.tan,                     (-1.35, 1.35), (-5, 5),    (0, 9),      "1 ÷ cos²x"),
    "sin⁻¹x":  (np.arcsin,                  (-0.97, 0.97), (-1.7, 1.7), (0, 4.5),   "1 ÷ √(1 − x²)"),
    "cos⁻¹x":  (np.arccos,                  (-0.97, 0.97), (0, 3.3),   (-4.5, 0),   "−1 ÷ √(1 − x²)"),
    "tan⁻¹x":  (np.arctan,                  (-6, 6),      (-1.7, 1.7), (0, 1.15),   "1 ÷ (1 + x²)"),
}
```

## 1 · The slope shadow

Pick a curve with the buttons, then drag the slider. The sage line is the tangent: it touches the curve at one
point and shows how steep the curve is there. Below, the slope is marked at the same x, and the marks join into the
shadow. The title names the shadow, the line of the table.

```python
def shadow(curve="x²", along=0.6):
    f, (x0, x1), ylim, slim, name = CURVES[curve]
    x = x0 + (x1 - x0) * along
    xs = np.linspace(x0, x1, 500)
    fig, (top, low) = plt.subplots(2, 1, figsize=(10, 7.2), sharex=True, gridspec_kw={"height_ratios": [1.2, 1]})
    top.plot(xs, f(xs), color=TEAL, lw=3, label=f"y = {curve}")
    s = slope(f, x)
    t = np.linspace(x - 0.9, x + 0.9, 2) if abs(s) < 4 else np.linspace(x - 0.3, x + 0.3, 2)
    top.plot(t, f(x) + s * (t - x), color=SAGE, lw=2.4, label="the tangent")
    top.plot([x], [f(x)], "o", color=CHALK, ms=8)
    top.set_ylim(*ylim); top.set_ylabel("the curve")
    top.set_title(f"y = {curve}   ·   at x = {x:.2f} the slope is {s:.3f}")
    done = np.linspace(x0, x, 300)
    low.plot(done, slope(f, done), color=SAGE, lw=3, label="its slope shadow")
    low.plot([x], [s], "o", color=CORAL, ms=10)
    low.axhline(0, color=DIM, lw=0.9)
    low.set_ylim(*slim); low.set_xlim(x0, x1); low.set_xlabel("x"); low.set_ylabel("its slope")
    low.set_title(f"the shadow: {name}")
    plt.tight_layout(); plt.show()

interact(shadow, curve=ToggleButtons(options=list(CURVES), value="x²", description=""),
         along=FloatSlider(value=0.0, min=0.0, max=1.0, step=0.05, description="along"));
```

**Notice.** A straight line casts a flat shadow, at its steepness. The parabola x² casts a straight line through
zero, **2x**: double where you are. eˣ casts itself. Sine casts cosine, and cosine casts minus sine.

## 2 · The step trick: x² and x³ from first principles

Take a small step h across, see how far the curve climbs, and divide by the step. For x² at 5 the climb divided by
the step is exactly 10 + h; for x³ at 2 it is 12 + 6h + h². As h shrinks, the slope settles.

```python
for h in (0.1, 0.01, 0.001):
    sq = ((5 + h) ** 2 - 25) / h
    cu = ((2 + h) ** 3 - 8) / h
    print(f"step h = {h:<6}  x² at 5: {sq:.6f}    x³ at 2: {cu:.6f}")
```

## 3 · The power rule: the power comes down, and the new power is one less

d/dx xⁿ = n xⁿ⁻¹. It works for powers that aren't whole numbers too: √x is x to the one half.

```python
for n, x in ((1, 3.0), (2, 3.0), (3, 3.0), (4, 3.0), (0.5, 4.0), (-1, 2.0)):
    print(f"n = {n:>4}:  at x = {x}  measured slope {slope(lambda t: t ** n, x):9.5f}   n xⁿ⁻¹ = {n * x ** (n - 1):9.5f}")
```

## 4 · eˣ is its own shadow; aˣ is itself times ln a

Slope divided by height is the same everywhere on aˣ: it's ln a. For 2ˣ that is 0.693 (the shadow sits below the
curve), for 3ˣ it is 1.099 (above), and for e it is exactly 1.

```python
for a in (2, math.e, 3, 10):
    f = lambda x, a=a: a ** x
    ratios = [slope(f, x) / f(x) for x in (-1.0, 0.0, 1.5)]
    print(f"a = {a:7.4f}:  slope ÷ height = {ratios[0]:.4f}, {ratios[1]:.4f}, {ratios[2]:.4f}    ln a = {math.log(a):.4f}")
```

## 5 · ln x: the mirror of eˣ, so its slope is one over where you are

Log base a of x is ln x divided by ln a, so its slope is 1 ÷ (x ln a). For base ten the factor is 1 ÷ ln 10 =
0.4343, the number Euler printed as 0.43429448 in 1755.

```python
for x in (0.5, 1.0, 4.0, 10.0):
    print(f"x = {x:>4}:  slope of ln x {slope(math.log, x):.5f}  (1/x = {1 / x:.5f})    "
          f"slope of log₁₀ x {slope(math.log10, x):.5f}  (1/(x ln 10) = {1 / (x * math.log(10)):.5f})")
```

## 6 · Sine and cosine: a point walking round a circle of radius one

Measure the angle in radians, the distance walked along the circle, and the point walks at speed one. It always
heads at right angles to the radius. Split that direction into up and across: up is cos θ, across is −sin θ. So
sine grows at the rate cos θ and cosine changes at the rate −sin θ.

```python
th = math.pi / 6
fig, ax = plt.subplots(figsize=(6.2, 6.2))
a = np.linspace(0, 2 * np.pi, 300)
ax.plot(np.cos(a), np.sin(a), color=DIM, lw=2)
px, py = math.cos(th), math.sin(th)
ax.plot([0, px], [0, py], color=CHALK, lw=2)
ax.plot([px, px], [0, py], color=TEAL, lw=4, label=f"height: sin θ = {py:.3f}")
ax.annotate("", xy=(px - 0.6 * py, py + 0.6 * px), xytext=(px, py), arrowprops=dict(arrowstyle="->", color=CORAL, lw=3))
ax.plot([px, px], [py, py + 0.6 * px], color=SAGE, lw=3, ls="--", label=f"up: cos θ = {px:.3f}")
ax.plot([px, px - 0.6 * py], [py + 0.6 * px] * 2, color=SAGE, lw=1.5, ls=":", label=f"across: −sin θ = {-py:.3f}")
ax.set_xlim(-1.3, 1.3); ax.set_ylim(-1.3, 1.3); ax.set_aspect("equal"); ax.legend(loc="lower left")
ax.set_title("θ = π/6: the direction of the walk (coral), at right angles to the radius")
plt.show()
```

**Notice.** The shadows go round like a clock: sin → cos → −sin → −cos → sin. A quarter turn and a sign each time.
In degrees every slope shrinks to π/180 ≈ 0.0175 of itself.

```python
for x in (0.0, math.pi / 6, 1.0):
    print(f"x = {x:.4f}:  slope of sin {slope(math.sin, x):+.5f} (cos {math.cos(x):+.5f})   "
          f"slope of cos {slope(math.cos, x):+.5f} (−sin {-math.sin(x):+.5f})")
print("in degrees, the slope of sine at 0:", round(slope(lambda d: math.sin(math.radians(d)), 0.0), 5), "= π/180")
```

## 7 · tan x, by the fraction rule

tan = sin ÷ cos. Bottom times the top's slope, minus top times the bottom's slope, over the bottom squared:
(cos·cos − sin·(−sin)) ÷ cos² = 1 ÷ cos² x, because cos² + sin² = 1. It is never less than one.

```python
for x in (0.0, math.pi / 4, math.pi / 3):
    print(f"x = {x:.4f}:  slope of tan {slope(math.tan, x):.5f}    1 ÷ cos²x = {1 / math.cos(x) ** 2:.5f}")
```

## 8 · The inverse functions: flip the graph, flip the slope

A flip across y = x swaps rise and run, so a slope s becomes 1 ÷ s. At 0.6, sine's slope is 0.8 (a 3-4-5
triangle), so the slope of sin⁻¹ is 1.25. sin⁻¹x + cos⁻¹x = π/2, so their slopes cancel. And tan⁻¹ casts a gentle
hill, 1 ÷ (1 + x²).

```python
for x in (0.0, 0.6, 2.0, 10.0):
    row = f"x = {x:>4}:  "
    if abs(x) < 1:
        row += f"sin⁻¹ {slope(math.asin, x):.4f} (1/√(1 − x²) = {1 / math.sqrt(1 - x * x):.4f})   cos⁻¹ {slope(math.acos, x):+.4f}   "
    row += f"tan⁻¹ {slope(math.atan, x):.4f} (1/(1 + x²) = {1 / (1 + x * x):.4f})"
    print(row)
```

## 9 · Rebuild the table

The right-hand column is hidden. Pick a line, type its slope in Python's way of writing (for example `-sin(x)`,
`1/x`, `2**x * log(2)`, `1/(1 + x**2)`; `log` is the natural log), and press **check**. A correct line fills in.
The last line, sec x, is the challenge from the video.

```python
from IPython.display import display
from ipywidgets import Button, Output, Text, VBox, HBox, Layout

LINES = [  # (the function, its slope as a function of x, the stretch of x to test on, the slope as the table writes it)
    ("k = 4",   lambda x: 0 * x,                       (-3, 3),     "0"),
    ("x",       lambda x: 1 + 0 * x,                   (-3, 3),     "1"),
    ("xⁿ",      None,                                  None,        "n xⁿ⁻¹"),
    ("eˣ",      np.exp,                                (-2, 2),     "eˣ"),
    ("2ˣ",      lambda x: 2.0 ** x * np.log(2),        (-2, 2),     "2ˣ ln 2"),
    ("ln x",    lambda x: 1 / x,                       (0.2, 5),    "1/x"),
    ("log₁₀ x", lambda x: 1 / (x * np.log(10)),        (0.2, 5),    "1/(x ln 10)"),
    ("sin x",   np.cos,                                (-3, 3),     "cos x"),
    ("cos x",   lambda x: -np.sin(x),                  (-3, 3),     "−sin x"),
    ("tan x",   lambda x: 1 / np.cos(x) ** 2,          (-1.2, 1.2), "1/cos²x = sec²x"),
    ("sin⁻¹x",  lambda x: 1 / np.sqrt(1 - x * x),      (-0.9, 0.9), "1/√(1 − x²)"),
    ("cos⁻¹x",  lambda x: -1 / np.sqrt(1 - x * x),     (-0.9, 0.9), "−1/√(1 − x²)"),
    ("tan⁻¹x",  lambda x: 1 / (1 + x * x),             (-3, 3),     "1/(1 + x²)"),
    ("sec x",   lambda x: np.sin(x) / np.cos(x) ** 2,  (-1.2, 1.2), "sec x tan x"),
]
NAMES = {"sin": np.sin, "cos": np.cos, "tan": np.tan, "exp": np.exp, "log": np.log, "ln": np.log, "sqrt": np.sqrt,
         "asin": np.arcsin, "acos": np.arccos, "atan": np.arctan, "sec": lambda v: 1 / np.cos(v), "pi": np.pi, "e": np.e}
SOLVED = {"xⁿ"}                                     # the power rule is on screen above: one line given

def matches(i, typed):
    """True when the typed slope equals the true one at a dozen points (numbers, not algebra)."""
    _, true, (lo, hi), _ = LINES[i]
    s = typed.replace("−", "-").replace("^", "**").replace("×", "*").replace("÷", "/").replace("√", "sqrt")
    xs = np.linspace(lo, hi, 13)
    try:
        got = np.array(eval(s, {"__builtins__": {}}, dict(NAMES, x=xs)), dtype=float) * np.ones_like(xs)
    except Exception:
        return None
    return bool(np.allclose(got, true(xs), rtol=1e-6, atol=1e-9))

def table():
    for name, _, _, answer in LINES:
        print(f"  d/dx {name:<8} =  {answer if name in SOLVED else '?'}")

pick = ToggleButtons(options=[n for n, _, _, _ in LINES if n not in SOLVED], value="cos x", description="")
yours = Text(value="", placeholder="its slope, e.g. -sin(x)", description="slope:", layout=Layout(width="380px"))
go, out = Button(description="check"), Output()

def on_check(_=None):
    i = [n for n, _, _, _ in LINES].index(pick.value)
    ok = matches(i, yours.value) if yours.value.strip() else None
    if ok: SOLVED.add(pick.value)
    out.clear_output()
    with out:
        if yours.value.strip():
            print(f"✓ d/dx {pick.value} = {yours.value}" if ok else
                  ("I can't read that yet: check the brackets." if ok is None else f"✗ not the slope of {pick.value}: try its picture"))
        table()

go.on_click(on_check)
display(VBox([pick, HBox([yours, go]), out]))
on_check()
```
