## 1 · The sliding ladder

The foot is x metres out and the top y metres up, with x² + y² = 6.5². Differentiate in time:
2x dx/dt + 2y dy/dt = 0, so **dy/dt = −x · (dx/dt) ÷ y**. Only then put in the moment's numbers. Drag the foot out.

```python
L, FOOT_SPEED = 6.5, 0.3                      # metres; metres a second

def ladder(foot=2.5):
    top = math.sqrt(L ** 2 - foot ** 2)
    dy = -foot * FOOT_SPEED / top
    fig, ax = plt.subplots(figsize=(6.4, 6))
    ax.plot([0, 0], [0, 7], color=DIM, lw=4); ax.plot([0, 7], [0, 0], color=DIM, lw=4)
    ax.plot([foot, 0], [0, top], color=TEAL, lw=5)
    ax.annotate("", xy=(foot + 0.9, 0.25), xytext=(foot, 0.25), arrowprops=dict(arrowstyle="-|>", color=SAGE, lw=2.4))
    ax.annotate("", xy=(0.3, top + 6 * dy), xytext=(0.3, top), arrowprops=dict(arrowstyle="-|>", color=CORAL, lw=2.4))
    ax.set_xlim(-0.5, 7); ax.set_ylim(-0.5, 7); ax.set_aspect("equal"); ax.set_xlabel("metres")
    ax.set_title(f"foot {foot:.2f} m out, top {top:.2f} m up:  the top falls at {-100 * dy:.1f} cm/s")
    plt.show()

interact(ladder, foot=FloatSlider(value=2.5, min=0.5, max=6.3, step=0.05, description="foot (m)"));
```

**Notice.** At 2.5 m out the top falls at 12.5 cm a second, slower than the foot. At 6 m out it falls at 72 cm a
second, more than twice as fast. A steady pull, and the top speeds up as it nears the ground.

## 2 · The filling cone

A cone, point down, 4 m deep with a top radius of 2 m; water flows in at 0.5 m³ a minute. Similar triangles give
r = h/2, so V = πh³/12 and dV/dt = (πh²/4) dh/dt.

```python
def cone(depth=1.0):
    rate = 0.5 / (math.pi * depth ** 2 / 4)
    fig, ax = plt.subplots(figsize=(6, 5.4))
    ax.plot([-2, 0, 2], [4, 0, 4], color=DIM, lw=2.4)
    ax.fill([-depth / 2, 0, depth / 2], [depth, 0, depth], color=TEAL, alpha=0.35)
    ax.plot([-depth / 2, depth / 2], [depth, depth], color=TEAL, lw=3)
    ax.set_xlim(-2.6, 2.6); ax.set_ylim(-0.2, 4.4); ax.set_aspect("equal"); ax.set_xlabel("metres")
    ax.set_title(f"depth {depth:.2f} m:  the level rises at {100 * rate:.0f} cm a minute")
    plt.show()

interact(cone, depth=FloatSlider(value=1.0, min=0.3, max=4.0, step=0.05, description="depth (m)"));
```

**Notice.** At 1 m deep the level climbs at about 64 cm a minute; at 2 m, about 16 cm a minute: a quarter of the
speed, because the surface is four times as wide in area.

## 3 · Rising and falling: f(x) = x³ − 3x² − 9x + 5

f′(x) = 3x² − 6x − 9 = 3(x − 3)(x + 1) is zero at −1 and 3. Test one number in each piece: f′(−2) = 15, f′(0) = −9,
f′(4) = 15. So f rises up to −1, falls from −1 to 3, and rises after 3.

```python
f = lambda x: x ** 3 - 3 * x ** 2 - 9 * x + 5
df = lambda x: 3 * x ** 2 - 6 * x - 9
xs = np.linspace(-3, 5, 400)
fig, (ax, bx) = plt.subplots(2, 1, figsize=(8, 6.4), sharex=True, gridspec_kw={"height_ratios": [2, 1]})
ax.plot(xs, f(xs), color=TEAL, label="f(x)")
ax.plot([-1, 3], [f(-1), f(3)], "o", color=CORAL, ms=9)
ax.annotate("local max 10", (-1, 10), xytext=(8, 8), textcoords="offset points")
ax.annotate("local min −22", (3, -22), xytext=(8, -16), textcoords="offset points")
ax.legend(loc="upper left")
bx.plot(xs, df(xs), color=SAGE, label="f′(x)"); bx.axhline(0, color=DIM, lw=1)
bx.fill_between(xs, df(xs), 0, where=df(xs) < 0, color=CORAL, alpha=0.25)
bx.legend(loc="upper left"); bx.set_xlabel("x")
plt.tight_layout(); plt.show()
for x0 in (-2, 0, 4): print(f"f′({x0}) = {df(x0)}")
```

## 4 · The open box

Cut a square of side x from each corner of a 24 cm × 9 cm card and fold up the sides. V = x(24 − 2x)(9 − 2x), and
V′ = 12(x − 2)(x − 9). Drag the cut from 0 up to 4.5 cm.

```python
V = lambda x: x * (24 - 2 * x) * (9 - 2 * x)

def box(cut=0.0):
    xs = np.linspace(0, 4.5, 300)
    fig, (ax, bx) = plt.subplots(1, 2, figsize=(11.5, 4.6), gridspec_kw={"width_ratios": [1.2, 1]})
    ax.plot([0, 24, 24, 0, 0], [0, 0, 9, 9, 0], color=DIM)
    for cx, cy in ((0, 0), (24 - cut, 0), (0, 9 - cut), (24 - cut, 9 - cut)):
        ax.fill([cx, cx + cut, cx + cut, cx], [cy, cy, cy + cut, cy + cut], color=CORAL, alpha=0.6)
    ax.plot([cut, 24 - cut, 24 - cut, cut, cut], [cut, cut, 9 - cut, 9 - cut, cut], color=TEAL, ls="--")
    ax.set_aspect("equal"); ax.set_xlim(-1, 25); ax.set_ylim(-1, 10); ax.set_title("the card, cm")
    bx.plot(xs, V(xs), color=TEAL); bx.plot([cut], [V(cut)], "o", color=CORAL, ms=10)
    bx.axvline(2, color=DIM, lw=0.8, ls=":")
    bx.set_xlabel("cut x (cm)"); bx.set_ylabel("volume (cm³)"); bx.set_ylim(0, 220)
    bx.set_title(f"cut {cut:.1f} cm:  volume {V(cut):.0f} cm³")
    plt.tight_layout(); plt.show()

interact(box, cut=FloatSlider(value=0.0, min=0, max=4.5, step=0.25, description="cut (cm)"));
```

**Notice.** The volume climbs to 200 cm³ at a cut of 2 cm, then falls away. At x = 9 the slope is zero too, but two
9 cm cuts can't fit along a 9 cm side.

## 5 · The tank that uses the least metal

A closed cylinder holding 1 m³ (1000 litres): πr²h = 1, so h = 1/(πr²) and the surface is S = 2πr² + 2/r. S′ = 0
gives r³ = 1/(2π), r ≈ 0.54 m, and then h = 2r. Drag the radius.

```python
S = lambda r: 2 * math.pi * r ** 2 + 2 / r

def tank(radius=0.3):
    h = 1 / (math.pi * radius ** 2)
    rs = np.linspace(0.2, 1.2, 300)
    fig, (ax, bx) = plt.subplots(1, 2, figsize=(11, 4.6), gridspec_kw={"width_ratios": [0.8, 1.2]})
    ax.add_patch(plt.Rectangle((-radius, 0), 2 * radius, h, fill=False, ec=TEAL, lw=3))
    ax.set_xlim(-1.3, 1.3); ax.set_ylim(0, 8.2); ax.set_aspect("equal"); ax.set_xlabel("metres")
    ax.set_title(f"r = {radius:.2f} m, h = {h:.2f} m")
    bx.plot(rs, [S(r) for r in rs], color=TEAL); bx.plot([radius], [S(radius)], "o", color=CORAL, ms=10)
    r0 = (1 / (2 * math.pi)) ** (1 / 3)
    bx.axvline(r0, color=DIM, lw=0.8, ls=":")
    bx.set_xlabel("radius r (m)"); bx.set_ylabel("surface (m²)"); bx.set_ylim(4, 12)
    bx.set_title(f"surface {S(radius):.2f} m²   (h ÷ 2r = {h / (2 * radius):.2f})")
    plt.tight_layout(); plt.show()

interact(tank, radius=FloatSlider(value=0.3, min=0.18, max=1.2, step=0.04, description="radius (m)"));
```

**Notice.** The surface is least, about 5.54 m², at r ≈ 0.54 m, where h ÷ 2r = 1: the tank is as tall as it is wide.

## 6 · Check your answer

Type any function of x and press **check**. The computer finds where its slope is zero (its flat points) and tests
each one: the second derivative first, and the signs of the slope on each side when that says nothing. You can write
`2x` for 2 × x and `^` for a power. Run the first cell below once (it sets up the checker), then the next.

```python
import sympy as sp
from sympy.parsing.sympy_parser import parse_expr, standard_transformations, implicit_multiplication_application, convert_xor
from IPython.display import display
from ipywidgets import Button, Output, HBox, VBox, Layout

x = sp.Symbol("x", real=True)
RULES = standard_transformations + (implicit_multiplication_application, convert_xor)

def read(s):
    s = s.replace("−", "-").replace("×", "*").replace("÷", "/").replace("√", "sqrt").replace("ln", "log")
    return parse_expr(s, local_dict={"x": x, "e": sp.E, "pi": sp.pi}, transformations=RULES)

def flat_points(function):
    try:
        g = read(function)
    except Exception:
        print("I can't read that yet: check the brackets."); return
    d1, d2 = sp.diff(g, x), sp.diff(g, x, 2)
    print(f"f(x) = {g}\nf′(x) = {sp.factor(d1)}")
    pts = sorted(sp.solveset(d1, x, domain=sp.S.Reals), key=float) if d1 != 0 else []
    if not pts: print("no flat points"); return
    for c in pts:
        k = d2.subs(x, c)
        if k < 0: verdict = "local maximum (f″ < 0: bending down)"
        elif k > 0: verdict = "local minimum (f″ > 0: bending up)"
        else:
            l, r = d1.subs(x, c - sp.Rational(1, 1000)), d1.subs(x, c + sp.Rational(1, 1000))
            verdict = ("local maximum" if l > 0 > r else "local minimum" if l < 0 < r else "a pause, not a turn") + " (f″ = 0: signs)"
        print(f"  x = {c}:  f = {sp.nsimplify(g.subs(x, c))},  f″ = {k}  →  {verdict}")

def checker(function):
    """A box and a check button; the flat points and their tests print under them."""
    function_box = Text(value=function, description="f(x) =", layout=Layout(width="380px"))
    check_button, answer = Button(description="check"), Output()
    def on_check(_=None):
        answer.clear_output()
        with answer: flat_points(function_box.value)
    check_button.on_click(on_check)
    display(VBox([HBox([function_box, check_button]), answer]))
    on_check()
```

```python
checker("x^3 - 3x^2 - 9x + 5")
```
