# Limits: why 0 ÷ 0 isn't the answer

Put x = 2 into (x² − 4) ÷ (x − 2) and you get 0 ÷ 0. Many of us write "the limit does not exist", and lose every
mark. The limit is 4: zero over zero is not an answer, it's a message that a factor is hiding.

```python
# The look: one style for every figure in this series of notebooks. Run this cell first.
import math
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, FloatLogSlider, Text

SLATE, SLATE2 = "#1f2b28", "#2a3a36"        # the slate ground
CHALK, DIM = "#ece6d6", "#93a097"           # chalk, and chalk smudge
CORAL = "#e3896b"                           # the one key thing in each figure
TEAL, SAGE, LILAC = "#6fb6bb", "#a0c27c", "#c9a0d3"   # the function · slopes · areas

plt.rcParams.update({
    "figure.facecolor": SLATE, "axes.facecolor": SLATE, "savefig.facecolor": SLATE,
    "text.color": CHALK, "axes.labelcolor": CHALK, "axes.edgecolor": DIM, "axes.titlecolor": CHALK,
    "xtick.color": DIM, "ytick.color": DIM, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#3b4c47", "grid.linewidth": 0.8,
    "font.size": 13, "axes.titlesize": 15, "axes.labelsize": 13, "legend.fontsize": 12,
    "legend.facecolor": SLATE2, "legend.edgecolor": DIM, "legend.labelcolor": CHALK, "legend.framealpha": 0.9,
    "lines.linewidth": 2.4, "figure.dpi": 100,
})

def hole(ax, x, y, color=CORAL, size=11):
    """An open circle: the one point the graph leaves out."""
    ax.plot([x], [y], "o", ms=size, mfc=SLATE, mec=color, mew=2.4, zorder=5)

def table(f, a, steps=(0.1, 0.01, 0.001), digits=5):
    """f near a, from the left and from the right."""
    print(f"{'x':>12} {'value':>14}")
    for h in steps: print(f"{a - h:12.4f} {f(a - h):14.{digits}f}")
    print(f"{'(x = ' + str(a) + ')':>12} {'never used':>14}")
    for h in reversed(steps): print(f"{a + h:12.4f} {f(a + h):14.{digits}f}")
```

## 1 · Zero over zero: what the fraction does near 2

A limit asks where a function is **heading** as x gets close to a number, never what happens **at** it. So put in
numbers near 2, from both sides, and watch.

```python
f = lambda x: (x * x - 4) / (x - 2)
table(f, 2)
```

```python
def near_two(distance=0.5):
    xs = np.linspace(0, 4, 401); xs = xs[np.abs(xs - 2) > 1e-9]
    fig, ax = plt.subplots(figsize=(8, 5))
    ax.plot(xs, f(xs), color=TEAL, label="(x² − 4) ÷ (x − 2)")
    hole(ax, 2, 4)
    for x in (2 - distance, 2 + distance):
        ax.plot([x], [f(x)], "o", color=CHALK, ms=8)
        ax.annotate(f"{f(x):.4f}", (x, f(x)), xytext=(10, -18), textcoords="offset points", color=CHALK)
    ax.set_xlim(0, 4); ax.set_ylim(1, 7); ax.set_xlabel("x"); ax.set_ylabel("value")
    ax.set_title(f"x = 2 ± {distance:g}: both sides close in on 4"); ax.legend(loc="upper left")
    plt.show()

interact(near_two, distance=FloatSlider(value=0.5, min=0.01, max=1.5, step=0.01, description="distance"));
```

**Notice.** Drag the distance towards zero. Both dots slide into the hole at (2, 4). The top and the bottom share
the factor x − 2: the fraction is (x − 2)(x + 2) ÷ (x − 2) = x + 2, everywhere except at 2 itself.

## 2 · The five moves, tried in order

1. **Try the number.** If the bottom isn't zero, the value is the limit.
2. **Find the hiding factor.** 0 ÷ 0 from a polynomial means both contain (x − a): factor, cancel, try again.
3. **Clean up the root.** A square root gets its partner: multiply top and bottom by it.
4. **Match a known shape.** sin x ÷ x → 1 (in radians), (eˣ − 1) ÷ x → 1, ln(1 + x) ÷ x → 1, (1 + x)^(1/x) → e.
5. **Look at the biggest power.** As x runs off to infinity, divide every term by the biggest power.

For a join between two pieces: the left limit, the right limit and the value must agree.

## 3 · Find the hiding factor: (x³ − 1) ÷ (x² − 1)

Top and bottom look alike, so it's tempting to say the limit at 1 is 1. Both are zero at 1, so both contain
x − 1. Cancelled, it's (x² + x + 1) ÷ (x + 1), and at 1 that is 3 ÷ 2.

```python
cube = lambda x: (x ** 3 - 1) / (x * x - 1)
table(cube, 1)
print("cancelled form at 1:", (1 + 1 + 1) / (1 + 1))
```

## 4 · Clean up the root: (√(9 + x) − 3) ÷ x

Multiply top and bottom by the partner √(9 + x) + 3. The top becomes (9 + x) − 9 = x, the x cancels, and what's left
is 1 ÷ (√(9 + x) + 3). At 0 that is 1/6 = 0.16667.

```python
root = lambda x: (np.sqrt(9 + x) - 3) / x
cleaned = lambda x: 1 / (np.sqrt(9 + x) + 3)
table(root, 0)
print("cleaned-up form at 0:", cleaned(0.0), "  (1/6 =", 1 / 6, ")")
```

## 5 · Match a known shape: sin x ÷ x, in radians and in degrees

sin x ÷ x heads for 1 only when x is in radians. In degrees the same shape heads for π/180 ≈ 0.01745. The rule
for every known shape: the angle inside must match what you divide by. So sin 5x ÷ 5x → 1, and sin 5x ÷ sin 2x → 5/2.

```python
def known_shape(x=0.5):
    xs = np.linspace(-1.5, 1.5, 601); xs = xs[np.abs(xs) > 1e-9]
    fig, ax = plt.subplots(figsize=(8, 4.6))
    ax.plot(xs, np.sin(xs) / xs, color=TEAL, label="sin x ÷ x, x in radians")
    ax.plot(xs, np.sin(np.radians(xs)) / xs, color=DIM, ls="--", label="sin x ÷ x, x in degrees")
    hole(ax, 0, 1)
    ax.plot([x], [math.sin(x) / x], "o", color=CHALK, ms=8)
    ax.set_ylim(-0.1, 1.2); ax.set_xlabel("x"); ax.legend(loc="center")
    ax.set_title(f"x = {x:g}:  sin x ÷ x = {math.sin(x) / x:.5f}")
    plt.show()

interact(known_shape, x=FloatSlider(value=0.5, min=0.01, max=1.5, step=0.01, description="x"));
```

```python
for x in (0.1, 0.01, 0.001):
    print(f"x = {x:<6}  sin x ÷ x = {math.sin(x) / x:.5f}   in degrees: {math.sin(math.radians(x)) / x:.5f}")
print("π/180 =", round(math.pi / 180, 5))
print("sin 5x ÷ sin 2x at x = 0.001:", round(math.sin(0.005) / math.sin(0.002), 5))
```

**Notice.** The solid curve closes in on 1. The dashed one, in degrees, lies almost flat near 0.017. The other known
shapes follow the same matching rule: (e^(3x) − 1) ÷ x → 3, ln(1 + 4x) ÷ x → 4 and (1 − cos 4x) ÷ x² → 8.

```python
for name, g, want in (("(e^(3x) − 1) ÷ x", lambda x: (math.exp(3 * x) - 1) / x, 3),
                      ("ln(1 + 4x) ÷ x", lambda x: math.log(1 + 4 * x) / x, 4),
                      ("(1 − cos 4x) ÷ x²", lambda x: (1 - math.cos(4 * x)) / x ** 2, 8)):
    print(f"{name:20s}", "  ".join(f"{g(x):.4f}" for x in (0.1, 0.01, 0.001)), f"  → {want}")
```

## 6 · The strangest shape: (1 + 2x)^(1/x)

The inside heads for 1, and one to any power is one. But the power 1/x grows without end: it's a tug of war. The
known shape (1 + x)^(1/x) heads for e ≈ 2.718, and (1 + 2x)^(1/x) = [(1 + 2x)^(1/(2x))]² heads for e² ≈ 7.389.

```python
def tug_of_war(x=0.5):
    xs = np.logspace(-4, 0, 300)
    fig, ax = plt.subplots(figsize=(8, 4.6))
    ax.semilogx(xs, (1 + 2 * xs) ** (1 / xs), color=TEAL, label="(1 + 2x)^(1/x)")
    ax.axhline(math.e ** 2, color=CORAL, lw=1.6, ls="--", label=f"e² ≈ {math.e ** 2:.3f}")
    ax.plot([x], [(1 + 2 * x) ** (1 / x)], "o", color=CHALK, ms=9)
    ax.invert_xaxis(); ax.set_ylim(3, 8); ax.set_xlabel("x (heading for 0, to the right)"); ax.legend(loc="lower left")
    ax.set_title(f"x = {x:.4g}:  (1 + 2x)^(1/x) = {(1 + 2 * x) ** (1 / x):.4f}")
    plt.show()

interact(tug_of_war, x=FloatLogSlider(value=0.5, base=10, min=-4, max=0, step=0.1, description="x"));
```

## 7 · Look at the biggest power: x runs off to infinity

(3x² − x + 2) ÷ (5x² + 4): divide every term by x². The small terms shrink to nothing and 3/5 is left. And
√(x² + 3x) − x isn't 0: with its partner it becomes 3x ÷ (√(x² + 3x) + x), which heads for 3/2.

```python
def at_infinity(x=10.0):
    xs = np.logspace(0, 4, 300)
    fig, ax = plt.subplots(figsize=(8, 4.6))
    ax.semilogx(xs, (3 * xs ** 2 - xs + 2) / (5 * xs ** 2 + 4), color=TEAL, label="(3x² − x + 2) ÷ (5x² + 4)")
    ax.semilogx(xs, np.sqrt(xs ** 2 + 3 * xs) - xs, color=LILAC, label="√(x² + 3x) − x")
    for v in (0.6, 1.5): ax.axhline(v, color=CORAL, lw=1.2, ls="--")
    ax.axvline(x, color=DIM, lw=1)
    ax.set_ylim(0, 2); ax.set_xlabel("x"); ax.legend(loc="center right")
    a1, a2 = (3 * x * x - x + 2) / (5 * x * x + 4), math.sqrt(x * x + 3 * x) - x
    ax.set_title(f"x = {x:,.0f}:  {a1:.4f}  and  {a2:.4f}   (→ 3/5 and 3/2)")
    plt.show()

interact(at_infinity, x=FloatLogSlider(value=10, base=10, min=0, max=4, step=0.1, description="x"));
```

## 8 · Closing a gap: a function in three pieces

Left of 3 the function is (x² − 9) ÷ (x − 3); at 3 it is a; right of 3 it is bx + 2. It is one unbroken curve when
the left limit, the right limit and the value agree: **lim_{x→3} f(x) = f(3)**. Find a and b with the sliders.

```python
def join(a=4.0, b=2.0):
    fig, ax = plt.subplots(figsize=(8, 4.8))
    xl = np.linspace(0, 2.999, 200); xr = np.linspace(3.001, 6, 200)
    ax.plot(xl, (xl ** 2 - 9) / (xl - 3), color=TEAL, label="left: (x² − 9) ÷ (x − 3)")
    ax.plot(xr, b * xr + 2, color=LILAC, label="right: bx + 2")
    hole(ax, 3, 6, color=TEAL); hole(ax, 3, 3 * b + 2, color=LILAC)
    ok = abs(a - 6) < 1e-9 and abs(3 * b + 2 - 6) < 0.02
    ax.plot([3], [a], "o", color=CORAL if ok else CHALK, ms=10, label=f"the value at 3: a = {a:g}")
    ax.set_xlim(0, 6); ax.set_ylim(0, 14); ax.legend(loc="upper left")
    ax.set_title("unbroken" if ok else f"left → 6, right → {3 * b + 2:.2f}, value {a:g}")
    plt.show()

interact(join, a=FloatSlider(value=4, min=0, max=10, step=0.5, description="a"),
         b=FloatSlider(value=2, min=0, max=3, step=1 / 3, description="b", readout_format=".3f"));
```

**Notice.** The left piece heads for 6, so a must be 6. The right piece gives 3b + 2, which is 6 when b = 4/3.

## 9 · L'Hôpital's rule (beyond CBSE)

For 0 ÷ 0 (or ∞ ÷ ∞) only: replace the top and the bottom by their own slopes, each on its own, and try again. The
rule is in l'Hôpital's 1696 textbook; Johann Bernoulli had sent it to him in a letter of 22 July 1694.
(x − sin x) ÷ x³ → (1 − cos x) ÷ 3x² → sin x ÷ 6x → cos x ÷ 6 → 1/6.

```python
for x in (0.1, 0.01, 0.001):
    print(f"x = {x:<6} (x − sin x) ÷ x³ = {(x - math.sin(x)) / x ** 3:.5f}")
print("1/6 =", round(1 / 6, 5))
```

## 10 · Check your answer

Type any limit: the function of x, and the number x goes to (use `oo` for infinity). Press **check**. The computer
works it out exactly, from both sides. Run the first cell below once (it sets up the checker), then the next. You can write `2x` for 2 × x and `^` for a power, and use sin, cos, tan,
exp, log (the natural log) and sqrt.

```python
import sympy as sp
from sympy.parsing.sympy_parser import parse_expr, standard_transformations, implicit_multiplication_application, convert_xor

X = sp.Symbol("x")
RULES = standard_transformations + (implicit_multiplication_application, convert_xor)
LAST = {}

def read(s):
    s = s.replace("−", "-").replace("×", "*").replace("÷", "/").replace("√", "sqrt").replace("∞", "oo").replace("ln", "log")
    return parse_expr(s, local_dict={"x": X, "e": sp.E, "pi": sp.pi, "oo": sp.oo}, transformations=RULES)

def check_your_answer(function="(x^2 - 4)/(x - 2)", goes_to="2"):
    try:
        fx, a = read(function), read(goes_to)
    except Exception:
        print("I can't read that yet: check the brackets."); return
    left, right = sp.limit(fx, X, a, "-"), sp.limit(fx, X, a, "+")
    LAST.update(f=fx, a=a)
    print(f"the limit, as x → {a}, of {fx}")
    if a in (sp.oo, -sp.oo) or left == right:
        lim = right if a != -sp.oo else left
        print(f"  = {lim}" + (f"  ≈ {float(lim):.4f}" if lim.is_real and lim.is_finite and not lim.is_Integer else ""))
    else:
        print(f"  from the left: {left},  from the right: {right}  → the two sides disagree: no limit")

from IPython.display import display
from ipywidgets import Button, Output, HBox, VBox, Layout

def checker(function, goes_to):
    """Two boxes and a check button; the answer prints under them."""
    function_box = Text(value=function, description="f(x) =", layout=Layout(width="380px"))
    goes_to_box = Text(value=goes_to, description="x →", layout=Layout(width="160px"))
    check_button, answer = Button(description="check"), Output()
    def on_check(_=None):
        answer.clear_output()
        with answer: check_your_answer(function_box.value, goes_to_box.value)
    check_button.on_click(on_check)
    display(VBox([HBox([function_box, goes_to_box, check_button]), answer]))
    on_check()
```

```python
checker("(x^2 - 4)/(x - 2)", "2")
```

**Watch it close in.** Run this cell after the one above: drag x towards the number it goes to, and watch the value
settle.

```python
def watch(distance=0.5):
    fx, a = LAST.get("f", read("(x^2 - 4)/(x - 2)")), LAST.get("a", 2)
    g = sp.lambdify(X, fx, "math")
    if a in (sp.oo, -sp.oo):
        x = float(np.sign(float(a))) / distance; xs = np.sign(float(a)) / np.logspace(-4, 0, 300)
    else:
        x = float(a) + distance; xs = float(a) + np.logspace(-4, 0, 300)
    ys = []
    for v in xs:
        try: ys.append(float(g(v)))
        except Exception: ys.append(np.nan)
    lim = sp.limit(fx, X, a)
    fig, ax = plt.subplots(figsize=(8, 4.4))
    ax.semilogx(np.logspace(-4, 0, 300), ys, color=TEAL)
    if lim.is_real and lim.is_finite: ax.axhline(float(lim), color=CORAL, lw=1.6, ls="--", label=f"the limit: {lim}")
    try: val = float(g(x))
    except Exception: val = float("nan")
    ax.plot([distance], [val], "o", color=CHALK, ms=10)
    ax.invert_xaxis(); ax.set_xlabel("how far x still is from its number (getting closer →)"); ax.legend(loc="lower left")
    ax.set_title(f"x = {x:.5g}:  value {val:.5f}")
    plt.show()

interact(watch, distance=FloatLogSlider(value=0.5, base=10, min=-3, max=0, step=0.1, description="x off by"));
```
