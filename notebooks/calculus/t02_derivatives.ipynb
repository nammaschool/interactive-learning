# Derivatives: the xˣ trap

Differentiate xˣ with the power rule and you get x · xˣ⁻¹ = xˣ, which says the slope at x = 2 is 4. Measure it, and
the slope is about 6.77. The true answer is the power rule's 4 plus the exponential rule's 2.77, added together.

```python
# The look: one style for every figure in this series of notebooks. Run this cell first.
import math
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, FloatLogSlider, Text

SLATE, SLATE2 = "#1f2b28", "#2a3a36"        # the slate ground
CHALK, DIM = "#ece6d6", "#93a097"           # chalk, and chalk smudge
CORAL = "#e3896b"                           # the one key thing in each figure
TEAL, SAGE, LILAC = "#6fb6bb", "#a0c27c", "#c9a0d3"   # the function · slopes · areas

plt.rcParams.update({
    "figure.facecolor": SLATE, "axes.facecolor": SLATE, "savefig.facecolor": SLATE,
    "text.color": CHALK, "axes.labelcolor": CHALK, "axes.edgecolor": DIM, "axes.titlecolor": CHALK,
    "xtick.color": DIM, "ytick.color": DIM, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#3b4c47", "grid.linewidth": 0.8,
    "font.size": 13, "axes.titlesize": 15, "axes.labelsize": 13, "legend.fontsize": 12,
    "legend.facecolor": SLATE2, "legend.edgecolor": DIM, "legend.labelcolor": CHALK, "legend.framealpha": 0.9,
    "lines.linewidth": 2.4, "figure.dpi": 100,
})

def slope(f, x, h=1e-6):
    """The slope of f at x, measured: rise over run across a tiny window either side."""
    return (f(x + h) - f(x - h)) / (2 * h)
```

## 1 · Measure the slope of xˣ at x = 2

Step from 2 to 2 + h, see how far the curve rises, and divide by the step. As the step shrinks, rise over run
settles on the slope. The power rule says 4.

```python
xx = lambda x: x ** x
for h in (0.1, 0.01, 0.001, 0.0001):
    print(f"step {h:<7} rise ÷ run = {(xx(2 + h) - xx(2)) / h:.4f}")
print("the power rule says:", 2 * 2 ** (2 - 1))
```

## 2 · The method: from the outside in

1. **Name the outside shape.** A sum, a product, a fraction, a power, or a function of something?
2. **Use that one rule.**
3. **Step one layer in, and repeat.**
4. **When x sits in two places**, take logs, or differentiate both sides.
5. **Check with a number**: a tiny window's rise over run, as in section 1.

## 3 · From the definition: f(x) = 1/x

f′(x) = lim_{h→0} [f(x + h) − f(x)] ÷ h. For 1/x, the top is −h ÷ [x(x + h)]. Divide by h, cancel, and only then let
h go to 0: −1/x².

```python
x0 = 2.0
for h in (0.1, 0.01, 0.001):
    print(f"h = {h:<6} [1/(x+h) − 1/x] ÷ h = {(1 / (x0 + h) - 1 / x0) / h:.6f}")
print("−1/x² at x = 2:", -1 / x0 ** 2)
```

## 4 · The rules, each checked with a number

Each line: the rule's answer at x = 0.7, and the slope measured with a tiny window. They should agree.

```python
x0 = 0.7
cases = [("product  x² sin x", lambda x: x * x * math.sin(x), lambda x: 2 * x * math.sin(x) + x * x * math.cos(x)),
         ("quotient sin x ÷ (1 + cos x)", lambda x: math.sin(x) / (1 + math.cos(x)), lambda x: 1 / (1 + math.cos(x))),
         ("chain    cos³(2x)", lambda x: math.cos(2 * x) ** 3, lambda x: -6 * math.cos(2 * x) ** 2 * math.sin(2 * x)),
         ("mixed    x e^(−x²)", lambda x: x * math.exp(-x * x), lambda x: math.exp(-x * x) * (1 - 2 * x * x)),
         ("inverse  tan⁻¹[2x ÷ (1 − x²)]", lambda x: math.atan(2 * x / (1 - x * x)), lambda x: 2 / (1 + x * x))]
for name, f, df in cases:
    print(f"{name:32s} rule: {df(x0):+.6f}   measured: {slope(f, x0):+.6f}")
```

## 5 · The chain rule as gears

In cos³(2x) the inside, 2x, moves twice as fast as x; the cosine passes that on, and the cube passes on what it
gets. Drag x and watch the three layers' rates multiply to the slope.

```python
def gears(x=0.4):
    u1, u2 = 2 * x, math.cos(2 * x)
    r1, r2, r3 = 2, -math.sin(u1), 3 * u2 ** 2
    xs = np.linspace(0, 1.6, 300)
    fig, ax = plt.subplots(figsize=(8, 4.4))
    ax.plot(xs, np.cos(2 * xs) ** 3, color=TEAL, label="y = cos³(2x)")
    m = r1 * r2 * r3
    ax.plot(xs, np.cos(2 * x) ** 3 + m * (xs - x), color=SAGE, lw=1.6, label=f"slope = {m:+.3f}")
    ax.plot([x], [np.cos(2 * x) ** 3], "o", color=CORAL, ms=9)
    ax.set_ylim(-1.2, 1.2); ax.legend(loc="lower left")
    ax.set_title(f"2  ×  ({r2:+.3f})  ×  {r3:.3f}  =  {m:+.3f}")
    plt.show()

interact(gears, x=FloatSlider(value=0.4, min=0, max=1.5, step=0.05, description="x"));
```

## 6 · Implicit: x² + xy + y² = 7 at (1, 2)

Differentiate both sides as they stand: 2x + (y + x dy/dx) + 2y dy/dx = 0. At (1, 2): 4 + 5 dy/dx = 0, so the slope
is −4/5.

```python
X, Y = np.meshgrid(np.linspace(-3.5, 3.5, 400), np.linspace(-3.5, 3.5, 400))
fig, ax = plt.subplots(figsize=(5.6, 5.6))
ax.contour(X, Y, X ** 2 + X * Y + Y ** 2, levels=[7], colors=[TEAL])
xs = np.linspace(-0.5, 2.5, 10)
ax.plot(xs, 2 - 0.8 * (xs - 1), color=SAGE, lw=1.8, label="tangent, slope −4/5")
ax.plot([1], [2], "o", color=CORAL, ms=9)
ax.set_aspect("equal"); ax.legend(loc="lower left"); ax.set_title("x² + xy + y² = 7")
plt.show()
```

## 7 · Parametric: x = t², y = t³

dx/dt = 2t and dy/dt = 3t², so dy/dx = 3t/2: time cancels. Drag t and watch the slope.

```python
def moving_point(t=2.0):
    ts = np.linspace(-2.5, 2.5, 400)
    fig, ax = plt.subplots(figsize=(7.4, 5))
    ax.plot(ts ** 2, ts ** 3, color=TEAL)
    x, y, m = t * t, t ** 3, 1.5 * t
    xs = np.linspace(x - 1.5, x + 1.5, 10)
    ax.plot(xs, y + m * (xs - x), color=SAGE, lw=1.8)
    ax.plot([x], [y], "o", color=CORAL, ms=10)
    ax.set_xlim(-0.5, 7); ax.set_ylim(-16, 16)
    ax.set_title(f"t = {t:.2f}:  point ({x:.2f}, {y:.2f}),  dy/dx = 3t/2 = {m:.2f}")
    plt.show()

interact(moving_point, t=FloatSlider(value=2, min=-2.5, max=2.5, step=0.1, description="t"));
```

## 8 · Logs: a product of four factors, and xˣ

For y = x(x + 1)(x + 2)(x + 3), ln y is a sum, so (1/y) dy/dx = 1/x + 1/(x + 1) + 1/(x + 2) + 1/(x + 3). At x = 1:
y = 24, and the slope is 24 × 25/12 = 50.

For y = xˣ, ln y = x ln x, so (1/y) dy/dx = ln x + 1, and dy/dx = xˣ(ln x + 1). At 2: 4(ln 2 + 1) ≈ 6.77, which is
the power rule's 4 plus the exponential rule's 4 ln 2 ≈ 2.77.

```python
many = lambda x: x * (x + 1) * (x + 2) * (x + 3)
print("slope of x(x+1)(x+2)(x+3) at 1:", round(slope(many, 1.0), 6), "  24 × 25/12 =", 24 * 25 / 12)
print(f"xˣ at 2: 4(ln 2 + 1) = {4 * (math.log(2) + 1):.4f}  =  4  +  {4 * math.log(2):.4f}")
```

```python
xs = np.linspace(1.2, 2.6, 200)
fig, ax = plt.subplots(figsize=(8, 4.6))
ax.plot(xs, xs ** xs, color=TEAL, label="y = xˣ")
for m, col, lab in ((4, DIM, "power rule's slope: 4"), (4 * math.log(2), LILAC, "exponential rule's: 2.77"),
                    (4 * (math.log(2) + 1), CORAL, "true slope: 6.77 = 4 + 2.77")):
    ax.plot(xs, 4 + m * (xs - 2), color=col, lw=1.8, ls="--" if col != CORAL else "-", label=lab)
ax.plot([2], [4], "o", color=CORAL, ms=9); ax.set_ylim(0, 12); ax.legend(loc="upper left")
ax.set_title("each wrong rule saw only half of the change")
plt.show()
```

## 9 · Check your answer

Type a function of x and your derivative, then press **check**. The computer differentiates the function exactly and
tells you whether your answer matches. You can write `2x` for 2 × x and `^` for a power, and use sin, cos, tan, exp,
log (the natural log), sqrt and atan (for tan⁻¹). Run the first cell below once (it sets up the checker), then the
next.

```python
import sympy as sp
from sympy.parsing.sympy_parser import parse_expr, standard_transformations, implicit_multiplication_application, convert_xor
from IPython.display import display
from ipywidgets import Button, Output, HBox, VBox, Layout

x = sp.Symbol("x", positive=True)
RULES = standard_transformations + (implicit_multiplication_application, convert_xor)
LAST = {}

def read(s):
    s = s.replace("−", "-").replace("×", "*").replace("÷", "/").replace("√", "sqrt").replace("ln", "log")
    return parse_expr(s, local_dict={"x": x, "e": sp.E, "pi": sp.pi}, transformations=RULES)

def check_your_answer(function, yours):
    try:
        f, g = read(function), read(yours)
    except Exception:
        print("I can't read that yet: check the brackets."); return
    df = sp.diff(f, x)
    LAST.update(f=f, g=g)
    print(f"the derivative of {f} is {sp.simplify(df)}")
    if sp.simplify(df - g) == 0: print("✓ your answer matches")
    else: print(f"✗ not equal: your {g} is a different function")

def checker(function, yours):
    """Two boxes and a check button; the verdict prints under them."""
    function_box = Text(value=function, description="f(x) =", layout=Layout(width="300px"))
    yours_box = Text(value=yours, description="yours:", layout=Layout(width="360px"))
    check_button, answer = Button(description="check"), Output()
    def on_check(_=None):
        answer.clear_output()
        with answer: check_your_answer(function_box.value, yours_box.value)
    check_button.on_click(on_check)
    display(VBox([HBox([function_box, yours_box, check_button]), answer]))
    on_check()
```

```python
checker("x^2 sin(x)", "2x sin(x) + x^2 cos(x)")
```

**Measure it with a window.** Run this cell after the one above: it measures the slope of the function you typed at
x = 2. Drag the window smaller and smaller, and watch rise over run settle on the true slope.

```python
def window(step=0.5):
    f, g = LAST.get("f", read("x^2 sin(x)")), LAST.get("g", read("2x sin(x) + x^2 cos(x)"))
    F, Gd = sp.lambdify(x, f, "math"), sp.lambdify(x, g, "math")
    a = 2.0
    m = (F(a + step) - F(a)) / step
    true = float(sp.diff(f, x).subs(x, a))
    xs = np.linspace(a - 1, a + 1.2, 300)
    fig, ax = plt.subplots(figsize=(8, 4.6))
    ax.plot(xs, [F(v) for v in xs], color=TEAL, label=f"y = {f}")
    ax.plot([a, a + step], [F(a), F(a + step)], "o", color=CHALK, ms=7)
    ax.plot(xs, F(a) + m * (xs - a), color=SAGE, lw=1.8, label=f"rise ÷ run = {m:.4f}")
    ax.axhline(F(a), color=DIM, lw=0.8)
    ax.set_ylim(min(F(v) for v in xs) - 1, max(F(v) for v in xs) + 1); ax.legend(loc="upper left")
    ax.set_title(f"window {step:.4g} at x = 2:  {m:.4f}   (true {true:.4f}, yours {Gd(a):.4f})")
    plt.show()

interact(window, step=FloatLogSlider(value=0.5, base=10, min=-4, max=0, step=0.1, description="window"));
```
