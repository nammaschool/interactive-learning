# π from an endless sum, two centuries before Newton: infinite series

Around 1400, Madhava of Sangamagrama, in Kerala, found that 1 − 1/3 + 1/5 − 1/7 + … adds up to exactly π/4, and that
a curve can be built from its height, its slope and its bend, one term at a time. (Beyond CBSE: AP Calculus BC,

```python
# The look: one style for every figure in this series of notebooks. Run this cell first.
import math
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, FloatSlider, ToggleButton

SLATE, SLATE2 = "#1f2b28", "#2a3a36"        # the slate ground
CHALK, DIM = "#ece6d6", "#93a097"           # chalk, and chalk smudge
CORAL = "#e3896b"                           # the one key thing in each figure
TEAL, SAGE, LILAC = "#6fb6bb", "#a0c27c", "#c9a0d3"   # the curve · slopes · the terms

plt.rcParams.update({
    "figure.facecolor": SLATE, "axes.facecolor": SLATE, "savefig.facecolor": SLATE,
    "text.color": CHALK, "axes.labelcolor": CHALK, "axes.edgecolor": DIM, "axes.titlecolor": CHALK,
    "xtick.color": DIM, "ytick.color": DIM, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#3b4c47", "grid.linewidth": 0.8,
    "font.size": 13, "axes.titlesize": 15, "axes.labelsize": 13, "legend.fontsize": 12,
    "legend.facecolor": SLATE2, "legend.edgecolor": DIM, "legend.labelcolor": CHALK, "legend.framealpha": 0.9,
    "lines.linewidth": 2.4, "figure.dpi": 100,
})

def sine_terms(x, m):
    """The first m terms of x − x³/3! + x⁵/5! − …, added up."""
    return sum((-1) ** j * x ** (2 * j + 1) / math.factorial(2 * j + 1) for j in range(m))

print("Ready. Run the next cell.")
```

## 1 · Building the sine curve, one term at a time

Near zero the sine curve has height 0 and slope 1, so the first guess is the line y = x. It hugs the curve, then
runs away. The bend is fixed by −x³/3!, then +x⁵/5!, and so on: each term is the finer correction the last one
needed. (x is in radians.)

```python
def build_sine(terms=1):
    xs = np.linspace(-2 * np.pi, 2 * np.pi, 800)
    fig, ax = plt.subplots(figsize=(10, 5.2))
    ax.plot(xs, np.sin(xs), color=TEAL, lw=6, alpha=0.55, label="sin x")
    ax.plot(xs, [sine_terms(v, terms) for v in xs], color=CORAL, lw=2.4, label=f"{terms} term{'s' if terms > 1 else ''}")
    ax.set_ylim(-2.2, 2.2); ax.set_xlim(-2 * np.pi, 2 * np.pi); ax.set_xlabel("x (radians)")
    ax.set_title(f"{terms} term{'s' if terms > 1 else ''}:  at x = 3 the sum is {sine_terms(3, terms):.4f},  sin 3 = {math.sin(3):.4f}")
    ax.legend(loc="lower left"); plt.show()

interact(build_sine, terms=IntSlider(value=1, min=1, max=8, step=1, description="terms"));
```

**Notice.** Every extra term wraps the curve further onto the wave. At x = 3 the sums go 3, −1.5, 0.525, 0.091,
0.145, 0.141, closing on sin 3 = 0.1411.

## 2 · The π cell: ten terms, and Madhava's correction

Times the sum 1 − 1/3 + 1/5 − … by four and it creeps towards π, painfully slowly: ten terms give only 3.04.
Madhava stopped early and added one small last piece. His best one, after n terms, is (n² + 1) ÷ (4n³ + 5n),
added or taken away in turn. Switch it on.

```python
def pi_cell(terms=10, correction=False):
    n = terms
    s = sum((-1) ** k / (2 * k + 1) for k in range(n))
    if correction: s += (-1) ** n * (n * n + 1) / (4 * n ** 3 + 5 * n)
    value = 4 * s
    right = 0
    while right < 15 and math.floor(value * 10 ** (right + 1)) == math.floor(math.pi * 10 ** (right + 1)): right += 1
    sums = [4 * sum((-1) ** k / (2 * k + 1) for k in range(j)) for j in range(1, n + 1)]
    fig, ax = plt.subplots(figsize=(10, 4.6))
    ax.plot(range(1, n + 1), sums, "o-", color=TEAL, ms=6, lw=1.4, label="4 × (1 − 1/3 + 1/5 − …)")
    ax.axhline(math.pi, color=CORAL, lw=1.6, label="π")
    if correction: ax.plot([n], [value], "o", color=CORAL, ms=12, label="with Madhava's correction")
    ax.set_xlabel("terms"); ax.set_ylim(2.5, 4.1); ax.legend(loc="upper right")
    ax.set_title(f"{n} terms{' + correction' if correction else ''}:  {value:.7f}   ({right} decimal place{'s' if right != 1 else ''} right)")
    plt.show()

interact(pi_cell, terms=IntSlider(value=10, min=1, max=40, step=1, description="terms"),
         correction=ToggleButton(value=False, description="Madhava's correction"));
```

**Notice.** Ten terms alone don't even get the first decimal. With the correction, the same ten terms give
3.1415927: six decimal places right. Without it you would need about two million terms.

## 3 · e to the x, from its slopes

Every slope of eˣ at zero is 1, so eˣ = 1 + x + x²/2! + x³/3! + … Put in x = 1 and watch e = 2.71828… appear.

```python
def exp_terms(terms=4):
    total = sum(1 / math.factorial(j) for j in range(terms))
    print(f"{terms} terms: {total:.7f}    e = {math.e:.7f}")

interact(exp_terms, terms=IntSlider(value=4, min=1, max=12, step=1, description="terms"));
```

## 4 · Check your answer: sin 0.1 to six decimal places

The textbook problem: the third term of sin 0.1 is less than a ten-millionth, too small to move the sixth decimal.
And a slip to avoid: the series works in radians, so 6° must first become π/30.

```python
import sympy as sp
x = sp.symbols("x")
print("sin x =", sp.series(sp.sin(x), x, 0, 8))
print("two terms at 0.1:", round(0.1 - 0.1 ** 3 / 6, 6), "  sin 0.1 =", round(math.sin(0.1), 6))
d = math.pi / 30
print("6 degrees = π/30 =", round(d, 6), "  two terms:", round(d - d ** 3 / 6, 6), "  sin 6° =", round(math.sin(d), 6))
```
