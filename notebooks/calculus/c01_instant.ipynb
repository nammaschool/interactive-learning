# What is speed at an instant? Shrinking the window

A ball rolls down Galileo's ramp. At exactly two seconds, how fast is it going? In one instant it moves no distance
in no time, and zero divided by zero means nothing. The way out: take a window of time, shrink it, and watch where
the average speeds are heading.

```python
# The look: one style for every figure in these notebooks. Run this cell first.
import math, warnings
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, FloatLogSlider

SLATE, SLATE2 = "#1f2b28", "#2a3a36"        # the slate ground
CHALK, DIM = "#ece6d6", "#93a097"           # chalk, and chalk smudge
CORAL = "#e3896b"                           # the one key thing in each figure
TEAL, SAGE, LILAC = "#6fb6bb", "#a0c27c", "#c9a0d3"   # the curve · slopes and rates · areas

warnings.filterwarnings("ignore", message=".*as float was deprecated.*")   # a browser-only notice, harmless
plt.rcParams.update({
    "figure.facecolor": SLATE, "axes.facecolor": SLATE, "savefig.facecolor": SLATE,
    "text.color": CHALK, "axes.labelcolor": CHALK, "axes.edgecolor": DIM, "axes.titlecolor": CHALK,
    "xtick.color": DIM, "ytick.color": DIM, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#3b4c47", "grid.linewidth": 0.8,
    "font.size": 13, "axes.titlesize": 15, "axes.labelsize": 13, "legend.fontsize": 12,
    "legend.facecolor": SLATE2, "legend.edgecolor": DIM, "legend.labelcolor": CHALK, "legend.framealpha": 0.9,
    "lines.linewidth": 2.4, "figure.dpi": 100,
})

def d(t):
    """Our ramp: the distance in metres is the time in seconds, squared (Galileo's law)."""
    return t * t

def average_speed(f, a, h):
    """The average speed over the window of time from a to a + h: distance gained ÷ time taken."""
    return (f(a + h) - f(a)) / h
```

## 1 · The average over a window

The ball starts at rest. After 1 second it has rolled 1 metre, after 2 seconds 4 metres, after 3 seconds 9 metres.
A **window** of time from 2 to 3 seconds takes the ball from 4 metres to 9 metres: 5 metres in 1 second, an
average of 5 metres a second.

On a graph of distance against time, that average is the **slope** of the line through the window's two points.

```python
print(f"{'window':>16} {'distance':>16} {'average speed':>16}")
for a, b in ((0, 1), (1, 2), (2, 3), (0, 3)):
    print(f"{a} s to {b} s".rjust(16), f"{d(a)} m to {d(b)} m".rjust(16), f"{average_speed(d, a, b - a):.0f} m/s".rjust(16))
```

**Notice.** The ball never has one speed: 1 metre in the first second, 5 in the third. It speeds up the whole way.

## 2 · Shrink the window

Drag the slider to make the window narrower. The window always starts at 2 seconds and lasts **h** seconds. The
coral line goes through the window's two points; its slope is the window's average speed. The dashed line is the
**tangent**, the one line that just touches the curve at 2 seconds.

```python
def shrink(h=1.0):
    a = 2.0
    s = average_speed(d, a, h)
    fig, (ax, bx) = plt.subplots(1, 2, figsize=(11.5, 5), gridspec_kw={"width_ratios": [1.5, 1]})
    t = np.linspace(0, 3.2, 400)
    ax.plot(t, d(t), color=TEAL, lw=3, label="distance = time²")
    tt = np.array([0.6, 3.2])
    ax.plot(tt, 4 + 4 * (tt - a), color=DIM, lw=1.6, ls="--", label="the tangent at 2 s (slope 4)")
    ax.plot(tt, d(a) + s * (tt - a), color=CORAL, lw=2.4, label=f"through the window (slope {s:.4f})")
    ax.plot([a, a + h], [d(a), d(a + h)], "o", color=CORAL, ms=8)
    ax.set_xlim(0.6, 3.2); ax.set_ylim(0, 10.5)
    ax.set_xlabel("time (s)"); ax.set_ylabel("distance (m)"); ax.legend(loc="upper left")
    ax.set_title(f"window: 2 s to {a + h:.4g} s")
    hs = np.logspace(0, -4, 60)
    bx.semilogx(hs, average_speed(d, a, hs), color=SAGE, lw=2.4)
    bx.plot([h], [s], "o", color=CORAL, ms=10)
    bx.axhline(4, color=DIM, lw=1.2, ls="--")
    bx.invert_xaxis(); bx.set_ylim(3.8, 5.2)
    bx.set_xlabel("window width h (s), shrinking →"); bx.set_ylabel("average speed (m/s)")
    bx.set_title(f"average speed = {s:.4f} m/s")
    plt.tight_layout(); plt.show()

interact(shrink, h=FloatLogSlider(value=1.0, base=10, min=-3, max=0, step=0.25, description="window h (s)",
                                  readout_format=".3g"));
```

**Notice.** As h shrinks, the coral line swings round onto the dashed tangent, and the average speed heads to 4.
No window ever divides zero by zero: every window gives an ordinary number.

```python
print(f"{'window width h':>16} {'average speed':>16}")
for k in range(0, 7):
    h = 10.0 ** -k
    print(f"{h:16.6f} {average_speed(d, 2, h):16.6f}")
print("heading to 4: at 2 seconds the ball is going 4 metres a second")
```

## 3 · Why exactly 4? The algebra

For a window of any width h: the distance goes from 4 to (2 + h)² = 4 + 4h + h². So the ball travels 4h + h² in a
time of h, and the average speed is (4h + h²) ÷ h = **4 + h**, exactly. As h shrinks, 4 + h heads to 4. We never
had to put h = 0.

```python
for h in (0.5, 0.1, 0.01, 0.001):
    print(f"h = {h:<6}  average = {average_speed(d, 2, h):.6f}   4 + h = {4 + h:.6f}")
```

## 4 · Every moment of the roll

Do the same at other moments: at 1 second the averages head to 2, at 3 seconds to 6. The speed is always **2 × the
time**: it grows by the same amount every second, which is Galileo's own definition of steady speeding up.

```python
def speed_at(a=2.0):
    s = average_speed(d, a, 1e-6)
    fig, ax = plt.subplots(figsize=(8, 4.8))
    t = np.linspace(0, 3.2, 400)
    ax.plot(t, d(t), color=TEAL, lw=3)
    tt = np.linspace(max(0, a - 1.2), min(3.2, a + 1.2), 2)
    ax.plot(tt, d(a) + s * (tt - a), color=CORAL, lw=2.4)
    ax.plot([a], [d(a)], "o", color=CORAL, ms=9)
    ax.set_xlim(0, 3.2); ax.set_ylim(0, 10.5); ax.set_xlabel("time (s)"); ax.set_ylabel("distance (m)")
    ax.set_title(f"at {a:.1f} s the speed is {s:.3f} m/s  (2 × {a:.1f} = {2 * a:.1f})")
    plt.show()

interact(speed_at, a=FloatSlider(value=2.0, min=0.2, max=3.0, step=0.1, description="time (s)"));
```

## 5 · Your own function

Run the cell, then type any function of x in the box (use `**` for powers, for example `x**3` or `x**2 + 3*x`)
and press **Enter**. The cell shrinks the window at x = 2 and prints where the slopes are heading.

```python
def slope_at(formula, a=2.0):
    names = {k: getattr(np, k) for k in ("sin", "cos", "tan", "exp", "log", "sqrt", "pi")}
    f = lambda x: eval(formula, {"__builtins__": {}}, dict(names, x=x))
    print(f"f(x) = {formula}, windows starting at x = {a:g}")
    for k in range(1, 6):
        h = 10.0 ** -k
        print(f"  width {h:<8} slope {average_speed(f, a, h):.6f}")
    s = average_speed(f, a, 1e-7)
    print(f"heading to about {s:.3f}")
    fig, ax = plt.subplots(figsize=(7, 4))
    x = np.linspace(a - 1.5, a + 1.5, 300)
    ax.plot(x, f(x), color=TEAL, lw=3)
    ax.plot(x, f(a) + s * (x - a), color=CORAL, lw=2, ls="--")
    ax.plot([a], [f(a)], "o", color=CORAL, ms=8)
    ax.set_title(f"slope at x = {a:g}: about {s:.3f}"); ax.set_xlabel("x")
    plt.show()

from ipywidgets import Text
interact(slope_at, formula=Text(value="x**2", description="f(x) =", continuous_update=False), a=FloatSlider(
    value=2.0, min=-3, max=3, step=0.5, description="at x ="));
```
