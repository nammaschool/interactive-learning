# The function that is its own slope: eˣ, ln x and sin x

Build a curve from one rule, "the rise is the height times the step", and it climbs to 2.718…, the number Jacob
Bernoulli met in 1683 through compound interest. That curve is eˣ: at every point its slope equals its height.
Its mirror image is ln x, whose slope is 1/x. And sine's slope is cosine only when angles are measured in radians.

```python
# The look: one style for every figure in this series of notebooks. Run this cell first.
import math
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider, SelectionSlider

SLATE, SLATE2 = "#1f2b28", "#2a3a36"        # the slate ground
CHALK, DIM = "#ece6d6", "#93a097"           # chalk, and chalk smudge
CORAL = "#e3896b"                           # the one key thing in each figure
TEAL, SAGE, LILAC = "#6fb6bb", "#a0c27c", "#c9a0d3"   # teal: the curve · sage: its slope · lilac: areas

plt.rcParams.update({
    "figure.facecolor": SLATE, "axes.facecolor": SLATE, "savefig.facecolor": SLATE,
    "text.color": CHALK, "axes.labelcolor": CHALK, "axes.edgecolor": DIM, "axes.titlecolor": CHALK,
    "xtick.color": DIM, "ytick.color": DIM, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#3b4c47", "grid.linewidth": 0.8,
    "font.size": 13, "axes.titlesize": 15, "axes.labelsize": 13, "legend.fontsize": 12,
    "legend.facecolor": SLATE2, "legend.edgecolor": DIM, "legend.labelcolor": CHALK, "legend.framealpha": 0.9,
    "lines.linewidth": 2.4, "figure.dpi": 100,
    "axes.prop_cycle": plt.cycler(color=[CHALK, TEAL, SAGE, LILAC, CORAL]),
})

def slope(f, x, h=1e-6):
    """The slope of f at x, measured over a tiny window: (f(x + h) − f(x − h)) ÷ 2h."""
    return (f(x + h) - f(x - h)) / (2 * h)
```

## 1 · Build a curve from its own rule

Start at height 1. Take a step across. The rule: **the rise is the height, times the step**. Then do it again from
the new height, and again, until you reach x = 1. With 10 steps of 0.1 you reach about 2.59. Use more, smaller
steps, and see where the height at x = 1 settles.

```python
def build(steps=10):
    dx = 1.0 / steps
    xs, ys = [0.0], [1.0]
    for _ in range(steps):
        ys.append(ys[-1] + ys[-1] * dx)          # rise = height × step
        xs.append(xs[-1] + dx)
    fig, ax = plt.subplots(figsize=(9, 5))
    x = np.linspace(0, 1, 200)
    ax.plot(x, np.exp(x), color=TEAL, lw=6, alpha=0.35, label="eˣ")
    ax.plot(xs, ys, "-o", color=CHALK, ms=4 if steps <= 50 else 0, lw=1.8, label=f"{steps} steps of {dx:g}")
    ax.plot([1], [ys[-1]], "o", color=CORAL, ms=11)
    ax.set_xlim(0, 1.05); ax.set_ylim(0.9, 2.9); ax.set_xlabel("x"); ax.set_ylabel("height")
    ax.set_title(f"after {steps} steps the height at x = 1 is {ys[-1]:.5f}")
    ax.legend(loc="upper left"); plt.show()

interact(build, steps=IntSlider(value=10, min=1, max=1000, step=1, description="steps"));
```

**Notice.** 10 steps give 2.59374, 100 give 2.70481, 1,000 give 2.71692. The answer doesn't run away: it settles
near **2.71828**, and the stepped line settles onto a smooth curve.

## 2 · Bernoulli's interest

In 1683 Jacob Bernoulli asked: lend 1 dollar at 100% a year, with the interest paid n times a year, each time on
what you have so far. What do you have after a year? It is (1 + 1/n)ⁿ, the same number as our steps.

```python
print(f"{'paid':>10} {'n':>9} {'after a year':>15}")
for name, n in (("once", 1), ("twice", 2), ("monthly", 12), ("daily", 365), ("hourly", 8760), ("each second", 31536000)):
    print(f"{name:>10} {n:9d} {(1 + 1 / n) ** n:15.6f}")
print(f"{'':>10} {'e':>9} {math.e:15.6f}")
```

**Notice.** The total grows with n, but it is stuck between 2 and 3. Euler called this number **e**, and in 1748
he gave it to 18 decimal places: 2.718281828459045235.

## 3 · The slope tracer: eˣ is its own slope

Drag the point along y = eˣ. The teal bar is the curve's **height** at the point. The sage line is its **slope**,
measured over a tiny window, and the sage step shows its rise for one step across. Every slope you measure is left
behind as a coral dot, placed at that x at the height of the slope.

```python
def tracer(x=-2.0):
    xs = np.linspace(-2, 2.05, 400)
    fig, ax = plt.subplots(figsize=(10, 5.6))
    ax.plot(xs, np.exp(xs), color=TEAL, lw=3, label="y = eˣ")
    done = np.arange(-2.0, x + 1e-9, 0.1)
    ax.plot(done, [slope(np.exp, d) for d in done], "o", color=CORAL, ms=7, label="the slope, measured at each x")
    h, s = math.exp(x), slope(np.exp, x)
    ax.plot([x, x], [0, h], color=TEAL, lw=8, alpha=0.7, solid_capstyle="butt")
    t = np.array([x - 1.2, x + 1.2])
    ax.plot(t, h + s * (t - x), color=SAGE, lw=2.2, label="the slope line")
    ax.plot([x, x + 1], [h, h], color=SAGE, lw=1.2, ls=":"); ax.plot([x + 1, x + 1], [h, h + s], color=SAGE, lw=4)
    ax.set_xlim(-2.2, 3.0); ax.set_ylim(0, 13); ax.set_xlabel("x"); ax.set_ylabel("y")
    ax.set_title(f"x = {x:+.1f}:   height {h:.4f}   ·   slope {s:.4f}")
    ax.legend(loc="upper left"); plt.show()

interact(tracer, x=FloatSlider(value=-2.0, min=-2, max=1.8, step=0.1, description="point x"));
```

**Notice.** Every coral dot lands exactly on the curve itself. The slope of eˣ, drawn as a curve, *is* eˣ:
**d/dx eˣ = eˣ**. The next cell measures it at a few points.

```python
for x in (-1.0, 0.0, 1.0, 2.0):
    print(f"x = {x:+.0f}:  height e^x = {math.exp(x):8.4f}   slope over a tiny window = {slope(np.exp, x):8.4f}")
```

## 4 · The mirror: ln x has slope 1/x

Flip y = eˣ across the line y = x and you get y = ln x, the natural log: every point (a, b) becomes (b, a). A
slope is rise over run, and the mirror swaps rise and run, so the slope at the mirror point is **one over** the old
slope. On eˣ the slope is the height, so on ln x the slope is one over x.

```python
def mirror(a=1.0):
    xs = np.linspace(-2.5, 2.2, 300)
    fig, ax = plt.subplots(figsize=(7, 7))
    ax.plot(xs, np.exp(xs), color=DIM, lw=2, label="y = eˣ")
    ln_x = np.linspace(0.06, 9, 400)
    ax.plot(ln_x, np.log(ln_x), color=TEAL, lw=3, label="y = ln x")
    ax.plot([-3, 9], [-3, 9], color=DIM, lw=1, ls="--")
    b = math.exp(a)
    ax.plot([a], [b], "o", color=DIM, ms=9); ax.plot([b], [a], "o", color=CORAL, ms=10)
    t = np.linspace(-1, 1, 2)
    ax.plot(a + t, b + b * t, color=DIM, lw=1.5)                     # slope b on eˣ
    ax.plot(b + t * 1.5, a + t * 1.5 / b, color=SAGE, lw=2.5)        # slope 1/b on ln x
    ax.set_xlim(-3, 9); ax.set_ylim(-3, 9); ax.set_aspect("equal")
    ax.set_title(f"on ln x at x = {b:.3f}: slope {slope(np.log, b):.4f} = 1/{b:.3f}")
    ax.legend(loc="lower right"); plt.show()

interact(mirror, a=FloatSlider(value=1.0, min=-1.5, max=2.0, step=0.1, description="point"));
```

**Notice.** At a = 1 the point (1, e) has slope e ≈ 2.718; its mirror (e, 1) has slope 1/e ≈ 0.368. And each
tenfold step in x adds the same height to ln x, about 2.303, while the slope there falls to a tenth.

```python
for x in (1, 10, 100, 1000):
    print(f"ln {x:>4} = {math.log(x):7.3f}    slope at {x:>4}: {slope(math.log, x):.5f}   (1/x = {1 / x:.5f})")
```

## 5 · Sine's slope, in radians and in degrees

Below is one full turn of sine, and its slope traced point by point (sage). Measured in radians a full turn is 2π,
about 6.283 units across, and the slope curve is exactly cosine: it starts at 1. Now switch the units to degrees,
360 across. The sine curve looks the same, but every unit across is now a much smaller turn, so every slope shrinks
by π/180.

```python
def units(full_turn=6.283):
    deg = full_turn > 100
    N = 360.0 if deg else 2 * math.pi
    f = (lambda x: np.sin(np.radians(x))) if deg else np.sin
    xs = np.linspace(0, N, 500)
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(xs, f(xs), color=TEAL, lw=3, label="sine")
    ax.plot(xs, slope(f, xs), color=SAGE, lw=3, label="its slope")
    ax.plot([0], [slope(f, 0.0)], "o", color=CORAL, ms=10)
    ax.axhline(0, color=DIM, lw=0.9)
    ax.set_xlim(0, N); ax.set_ylim(-1.25, 1.25)
    ax.set_xlabel("angle, in " + ("degrees" if deg else "radians"))
    ax.set_title(f"in {'degrees' if deg else 'radians'}: the slope at 0 is {slope(f, 0.0):.5f}"
                 + ("  (= π/180)" if deg else ""))
    ax.legend(loc="lower left"); plt.show()

interact(units, full_turn=SelectionSlider(options=[("6.283", 6.283), ("360", 360.0)], value=6.283,
                                          description="a full turn"));
```

**Notice.** In radians the slope at 0 is 1 and the slope curve is cosine. In degrees it is 0.01745, so the slope of
sine is (π/180) × cosine. Radians are the one unit where that extra number disappears. That is also why every
programming language's sine expects radians:

```python
print("sin(π/6) =", round(math.sin(math.pi / 6), 6), "   sin(30) =", round(math.sin(30), 3), "  (30 radians!)")
```

## 6 · Mirrors give the inverse functions

Flip sine (from −π/2 to π/2) across y = x and you get inverse sine, written sin⁻¹. The mirror turns the slope
cos y into 1/cos y, and cos y = √(1 − x²). So **d/dx sin⁻¹x = 1/√(1 − x²)**. The same flip on tangent, whose slope
is 1 + tan² y, gives **d/dx tan⁻¹x = 1/(1 + x²)**.

```python
for x in (0.0, 0.5, 0.9):
    print(f"x = {x}:  slope of sin⁻¹ {slope(np.arcsin, x):.4f}   1/√(1 − x²) = {1 / math.sqrt(1 - x * x):.4f}   |   "
          f"slope of tan⁻¹ {slope(np.arctan, x):.4f}   1/(1 + x²) = {1 / (1 + x * x):.4f}")
```

## 7 · The slope of the slope: sine bends back like a spring

Sine's slope is cosine, and cosine's slope is minus sine. So the **second derivative** of sine is sine, upside
down: y″ = −y. Wherever sine is high it bends down; wherever it is low it bends up.

```python
x = np.linspace(0, 4 * np.pi, 600)
fig, axs = plt.subplots(3, 1, figsize=(10, 6.4), sharex=True)
for ax, (y, name, col) in zip(axs, ((np.sin(x), "y = sin x", TEAL), (np.cos(x), "y′ = cos x", SAGE),
                                     (-np.sin(x), "y″ = −sin x", SAGE))):
    ax.plot(x, y, color=col); ax.set_ylim(-1.3, 1.3); ax.set_ylabel(name, rotation=0, ha="right", va="center")
axs[2].plot(x, np.sin(x), color=CORAL, lw=1.2, ls="--", label="y, for comparison"); axs[2].legend(loc="upper right")
plt.tight_layout(); plt.show()
```

## 8 · x to the x: logarithmic differentiation

In y = xˣ the base and the power both move, so neither the power rule nor the eˣ rule applies. Take the log of
both sides: ln y = x ln x. Take the slope of each side: (1/y) y′ = ln x + 1. So **y′ = xˣ (ln x + 1)**. Drag the
point and compare the true slope with the power-rule slip, x · xˣ⁻¹.

```python
def xx(x=2.0):
    f = lambda t: t ** t
    xs = np.linspace(0.05, 2.6, 400)
    fig, ax = plt.subplots(figsize=(10, 5.4))
    ax.plot(xs, f(xs), color=TEAL, lw=3, label="y = xˣ")
    true, slip = slope(f, x), x * x ** (x - 1)
    t = np.array([x - 0.6, x + 0.6])
    ax.plot(t, f(x) + true * (t - x), color=SAGE, lw=2.5, label=f"true slope {true:.3f}")
    ax.plot(t, f(x) + slip * (t - x), color=DIM, lw=1.5, ls="--", label=f"power-rule slip {slip:.3f}")
    ax.plot([x], [f(x)], "o", color=CORAL, ms=10)
    ax.set_xlim(0, 2.7); ax.set_ylim(0, 12); ax.set_xlabel("x"); ax.set_ylabel("y")
    ax.set_title(f"x = {x:.2f}:  xˣ(ln x + 1) = {f(x) * (math.log(x) + 1):.3f}")
    ax.legend(loc="upper left"); plt.show()

interact(xx, x=FloatSlider(value=2.0, min=0.1, max=2.5, step=0.05, description="point x"));
```

**Notice.** At x = 2 the slip says 4, but the true slope is about 6.77. The power rule only works when the power
stays still. Somewhere on the left the curve is flattest, with a slope of zero: that's the challenge in the video.

## Check the rules with sympy

sympy does the algebra of slopes exactly. Run the cell: each line is a rule from the video.

```python
import sympy as sp
X = sp.symbols("x", positive=True)
for f in (sp.exp(X), sp.log(X), sp.sin(X), sp.asin(X), sp.atan(X), X ** X):
    print(f"d/dx {str(f):10} = {sp.simplify(sp.diff(f, X))}")
print(f"second derivative of sin(x) = {sp.diff(sp.sin(X), X, 2)}")
```
