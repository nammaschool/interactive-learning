## 1 · The trick

Stand a stick straight up. The Sun makes a shadow of the stick and a shadow of the tree at the same moment, so the
sunlight hits both at the same angle. That makes two triangles with the same shape, just different sizes.

Same shape means the same ratio:

**height ÷ shadow is the same for the stick and for the tree.**

So: tree height = tree shadow × (stick height ÷ stick shadow).

```python
stick_height = 1.0    # metres
stick_shadow = 1.6    # metres, measured on the ground
tree_shadow  = 12.0   # metres

ratio = stick_height / stick_shadow
tree_height = tree_shadow * ratio
print(f"height ÷ shadow = {ratio:.3f}")
print(f"so the tree is about {tree_height:.1f} m tall")
```

## 2 · See the two triangles

```python
def draw(stick_height=1.0, stick_shadow=1.6, tree_shadow=12.0):
    tree_height = tree_shadow * stick_height / stick_shadow
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.fill([0, tree_shadow, 0], [0, 0, tree_height], alpha=0.25, color="tab:green", label="tree and its shadow")
    x0 = tree_shadow + 2
    ax.fill([x0, x0 + stick_shadow, x0], [0, 0, stick_height], alpha=0.5, color="tab:orange", label="stick and its shadow")
    ax.annotate(f"tree: {tree_height:.1f} m", (0, tree_height / 2), xytext=(6, 0), textcoords="offset points")
    ax.annotate(f"stick: {stick_height:.1f} m", (x0, stick_height), xytext=(0, 6), textcoords="offset points")
    ax.set_aspect("equal"); ax.set_xlabel("metres"); ax.set_ylabel("metres"); ax.grid(alpha=0.3); ax.legend(loc="upper right")
    plt.show()

interact(draw, stick_height=FloatSlider(value=1.0, min=0.5, max=2.0, step=0.1),
         stick_shadow=FloatSlider(value=1.6, min=0.2, max=5.0, step=0.1),
         tree_shadow=FloatSlider(value=12.0, min=2.0, max=30.0, step=0.5));
```

## 3 · The ratio only depends on the Sun's angle

Move the Sun. For each angle, every upright thing has the same height ÷ shadow, whether it is a stick, a tree or a
tower. Mathematicians gave that ratio a name: the **tangent** of the angle.

```python
def sun(angle_deg=32.0):
    a = math.radians(angle_deg)
    things = (("stick", 1.0, "tab:orange"), ("person", 1.7, "tab:blue"), ("tree", 6.0, "tab:green"))
    fig, axes = plt.subplots(1, 3, figsize=(10, 3.6), sharey=True)
    for ax, (name, h, colour) in zip(axes, things):
        s = h / math.tan(a)
        ax.plot([0, 0], [0, h], color=colour, lw=4)                      # the thing, standing up
        ax.plot([0, s], [0, 0], color="0.3", lw=6, alpha=0.5)            # its shadow
        ax.plot([0, s], [h, 0], color=colour, ls="--", lw=1)             # the sunlight past its top
        ax.set_title(f"{name}: {h:g} ÷ {s:.2f} = {h / s:.3f}", fontsize=10)
        ax.set_xlim(-0.5, max(2.0, 1.15 * s)); ax.set_xlabel("shadow (m)"); ax.grid(alpha=0.3)
    axes[0].set_ylabel("height (m)"); axes[0].set_ylim(0, 6.6)
    fig.suptitle(f"Sun at {angle_deg:.0f}°: height ÷ shadow = {math.tan(a):.3f} for all three")
    plt.tight_layout(); plt.show()

interact(sun, angle_deg=FloatSlider(value=32.0, min=10.0, max=80.0, step=1.0));
```

## 4 · A table you can use on any sunny day

Write the ratio down once for each angle, and you never need the stick again: measure the angle of the Sun and the
shadow, and look up the ratio. Tables like this were the first tools of trigonometry.

```python
for angle in range(10, 81, 10):
    print(f"{angle:2d}°   height ÷ shadow = {math.tan(math.radians(angle)):.3f}")
```
